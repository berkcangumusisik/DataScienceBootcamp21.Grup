### Makine Öğrenmesi Nedir?

Makine öğrenmesi, bilgisayarların insanlara benzer şekilde öğrenmesini sağlamak amacıyla çeşitli algoritma ve tekniklerin geliştirilmesi üzerine çalışan bilimsel bir araştırma alanıdır.

Bir başka deyişle; makine öğrenmesi, bilgisayarların *açıkça programlanmadan*, yani her adım için kural yazılmadan, verilerden örüntüler ve ilişkiler bularak yeni durumlar için tahmin, sınıflandırma veya öneride bulunabilmesini sağlar.

> Bilgisayarların insanlara benzer şekilde öğrenmesini sağlamak maksadıyla çeşitli algoritma ve tekniklerin geliştirilmesi için çalışılan bilimsel çalışma alanıdır.

Yani, makine öğrenmesi insan öğrenmesini taklit etmeye çalışır. Bunu yapabilmek için farklı algoritmalar ve yöntemler geliştirilir ve bilgisayarların geçmiş deneyimlerden yola çıkarak yeni durumları çözebilmesi amaçlanır.

## Basit Bir Tanım

Makine öğrenmesi, bilgisayarların verilerdeki örüntüleri (pattern) bulmayı ve bu bilgiyi yeni veriler üzerinde kullanmayı öğrenmesini sağlayan bir teknolojidir.

## Geleneksel Programlama ile Farkı Nedir?

- **Geleneksel programlamada:** Kurallar (“If-Else” gibi) elle yazılır ve bilgisayar yalnızca verilen komutları uygular.
- **Makine öğrenmesinde:** Kurallar ve örüntüler bilgisayar tarafından *veriler kullanılarak* otomatik olarak öğrenilir.

## Örneklerle Makine Öğrenmesi

- **E-posta spam filtresi:** Sistem, geçmişte spam veya spam olmayan e-postalara bakarak yeni gelen bir e-postanın spam olup olmadığını tahmin eder.
- **Film önerileri:** İzlediğiniz filmler ve diğer kullanıcıların tercihleri üzerinden size yeni filmler önerilir.
- **El yazısı tanıma:** Telefonlarda kendi el yazınızı tanıyan sistemler makine öğrenmesiyle çalışır.

## Makine Öğrenmesinin Temel Amacı

Makine öğrenmesinin ana amacı, geçmişten öğrenerek yeni durumlar için *otomatik tahmin* yapabilen veya karar alabilen sistemler geliştirmektir.

## En Temel Kavramlar

- **Veri:** Makine öğrenmesinin yakıtı veridir. Algoritmalar, çok miktarda veriyi analiz ederek örüntüleri bulur.
- **Model:** Öğrenilen bilginin bilgisayarda temsil edilmesi.
- **Eğitim (Training):** Modelin verilerden öğrenme sürecidir.
- **Tahmin (Prediction):** Eğitilmiş modelin yeni verilerde sonuç üretmesi.

---

Sonuç olarak, makine öğrenmesi: bilgisayarları adım adım programlamak yerine onları *örnekler üzerinden nasıl karar vereceklerini* öğretebilmemizi sağlayan bir yaklaşımdır. Böylece, bilgisayarlar insanlar gibi deneyimlerden öğrenip yeni durumlarda otomatik olarak karar alabilir hale gelir.


### Makine Öğrenmesinde Değişken Türleri (Hiç Bilmeyenler İçin)

Makine öğrenmesi, bilgisayarların verilerden anlamlar çıkarmasını sağlar. Ancak bilgisayara verilen verilerin yapısı yani **değişken türleri** (variable types) çok önemlidir. Değişken, kısacası bir veri setinde her satıra ait bir özelliktir. Örneğin bir kişinin yaşı, cinsiyeti, kilosu birer değişkendir.

Değişken türlerini anlamak, doğru analiz yapmak ve uygun yöntemleri kullanmak için ilk adımdır. Şimdi hiç bilmeyen biri için temel türleri sade şekilde açıklayalım:

---

#### 1. Sayısal Değişkenler (Numerik Variables)

Bu değişkenler rakamsal (sayılarla ifade edilen) verilerdir ve onlarla matematiksel işlemler yapılabilir.

- **Sürekli (Continuous):** Virgüllü (onlu) veya çok geniş aralıklarda sayı değeri alabilir.
  - *Örnekler:* Boy (170.5 cm), kilo (72.8 kg), aylık gelir (4580 TL), hava sıcaklığı (18.9°C)

- **Kesikli (Discrete):** Genelde tam sayı olur, sayıların arası atlamalıdır (1, 2, 3 gibi).
  - *Örnekler:* Kardeş sayısı (0, 1, 2...), oda sayısı (3), sipariş adedi (7)

#### 2. Kategorik Değişkenler (Categorical Variables)

Bu tür değişkenler sayı gibi işlemlere girmez; verileri kategori, grup veya etiket olarak tutar.

- **Nominal (Sınıflandırıcı):** Sadece grup bildirir; sıralama veya büyüklük ilişkisi yoktur.
  - *Örnekler:* Cinsiyet (Kadın, Erkek), şehir adı, saç rengi (siyah, sarı, kahverengi)

- **Ordinal (Sıralı):** Grupların arasında sıralama vardır ama aralarındaki 'mesafe' bilinmez.
  - *Örnekler:* Eğitim seviyesi (ilkokul < lise < üniversite), müşteri memnuniyeti (kötü, orta, iyi)

#### 3. İkili/Binary/Boolean Değişkenler

Sadece iki farklı değer alabilen değişkenlerdir. Genellikle "Evet/Hayır", "0/1" veya "Doğru/Yanlış" ile gösterilir.
  - *Örnekler:* Evli mi? (Evet/Hayır), Satın alma yaptı mı? (1 = yaptı, 0 = yapmadı)

#### 4. Tarih/Zaman Değişkenleri

Bilgi olarak bir olayın zamanını veya tarihini belirtirler.
  - *Örnekler:* Doğum tarihi (15.02.1990), sipariş saati (14:30), mezuniyet yılı (2020)

---

#### Değişkenlerin Diğer Temel Sınıflamaları

- **Bağımlı Değişken (target, dependent, output, response):** Modelin tahmin etmeye çalıştığı sonuçtur. Mesela hasta mı değil mi? müşterinin geliri ne olacak? gibi.

- **Bağımsız Değişken (feature, independent, input, column, predictor, explanatory):** Sonucu etkileyen, modele giriş olarak verilen diğer bilgiler (örneğin yaş, cinsiyet, alışveriş sıklığı...)

---

> **Küçük Not:** Değişken tiplerini bilmek, uygun analiz yöntemini seçmek, verideki hataları bulmak, eksik/veri tipi yanlışlarını düzeltmek ve doğru modellemeler yapmak için gereklidir.


## Makine Öğrenmesi Türleri

Makine öğrenmesi, bilgisayarlara veriler üzerinden otomatik öğrenme yeteneği veren bir teknoloji türüdür. Yani, bilgisayara bir şeyleri "programla" deyip adım adım anlatmak yerine, ona örnekler veririz ve kendi kurallarını çıkarıp tahmin yapmasını isteriz. Tıpkı bir çocuğun deneyerek öğrenmesi gibi.

Makine öğrenmesinde üç ana öğrenme türü bulunur. Bunları anlamak için günlük hayatta kullanılan örneklerle düşünebiliriz:

---
### 1. Denetimli Öğrenme (Supervised Learning)

**Bu ne demek?** Bilgisayara verdiğimiz örneklerin "doğru cevaplarını" (etiketlerini) da gösteriyoruz. O da bu doğru cevapları görerek öğreniyor. Yani: Hem soruyu hem cevabı biliyor!

**Gündelik Benzetme:** Bir çocuğa matematik problemi çözdürüyoruz ve her seferinde doğru cevabı söylüyoruz. Zamanla çocuk nasıl yapacağını öğreniyor ve yeni bir soru gelince cevabını tahmin ediyor.

**Örnekler:**
- Elinizde bin tane e-posta var ve hangisinin spam (istenmeyen) olduğunu işaretlediniz. Bilgisayar bu örneklerden öğrenerek gelen yeni bir e-postanın spam olup olmadığını tahmin etmeye çalışır.
- Hastaların yaş, kilo, tahlil gibi bilgilerini ve hastalığı olup olmadığını verdiniz. Bilgisayara yeni bir hastanın verisini verdiğinizde, hasta mı diye tahmin yapar.

--- 

### 2. Denetimsiz Öğrenme (Unsupervised Learning)

**Nedir?** Bu türde bilgisayara sadece verileri veririz; hangi verinin hangi gruba ait olduğunu, yani doğru cevabı söylemeyiz. Bilgisayar kendi başına verilerin arasında benzerliklere göre gruplar (kümeler, desenler) bulmaya çalışır.

**Gündelik Benzetme:** Hiç müzik türleri bilmeyen birine yüzlerce farklı şarkı dinletiyorsunuz, "bunları benzerlerine göre ayır" diyorsunuz ama jazz mı rock mı olduğunu söylemiyorsunuz. O, sese göre kendi kümelerini oluşturur.

**Örnekler:**
- Bir markette alışveriş yapan binlerce müşteriyi veri olarak veriyoruz. Bilgisayar benzer alışkanlıkları olanları grupluyor (ör. "çocuklu aileler", "öğrenciler" gibi).
- Elinizde pek çok belge var ve onları konularına göre ayırmak istiyorsunuz ama hangi konuda olduklarını bilmiyorsunuz.

---

### 3. Pekiştirmeli Öğrenme (Reinforcement Learning)

**Nedir?** Burada bilgisayar bir oyun gibi çevreye hareketler (eylemler) yapar ve sonucunda ödül veya ceza alır. Zamanla hangi hareket doğru, hangisi yanlış öğrenir ve en çok ödül getiren stratejiyi geliştirir.

**Gündelik Benzetme:** Bir köpeğe "otur" komutu öğretiyorsunuz. Doğru yaparsa ödül maması veriyorsunuz; yanlış yaparsa ödül yok. Köpek doğru davranışı zamanla öğreniyor.

**Örnekler:**
- Bilgisayara satranç veya video oyunu oynamayı öğretmek: Her hamlede kazanırsa ödül, kaybederse ceza.
- Otonom (sürücüsüz) arabaların trafikte doğru yolu bulmayı öğrenmesi: Kurallara uygun hareket ettiğinde ödül, hata yaptığında ceza.

---

## Model Başarı Değerlendirme Yöntemleri

Makine öğrenmesi modellerinin ne kadar iyi performans gösterdiğini anlamak için çeşitli başarı değerlendirme (model değerlendirme, model doğrulama) yöntemleri kullanılır. Temel amaç; modelin gerçek dünyadaki yeni verilere ne kadar doğru tahmin yapabileceğini ölçmektir. 

Başlıca değerlendirme yöntemleri ve metrikleri şunlardır:

### 1. Eğitim-Test Ayrımı (Train-Test Split)
- Veri seti genellikle ikiye ayrılır: **Eğitim seti** (modeli eğitmek için), **test seti** (başarıyı ölçmek için).
- Model önce eğitim verisinde öğrenir, sonra test verisinde hiç görmediği örnekler üzerinde performansı ölçülür.

### 2. Doğrulama Yöntemleri (Validation Techniques)
- **K-katlı Çapraz Doğrulama (K-Fold Cross-Validation):** Veri seti k eşit parçaya ayrılır; her seferinde bir parça test, kalanları eğitim için kullanılır. Tüm katlarda elde edilen sonuçlar ortalanır.
- **Leave-One-Out (LOO):** Her seferinde tek bir örnek test, geri kalanı eğitim için kullanılır.

### 3. Başarı Metrikleri
Metrik seçimi, probleme (sınıflandırma, regresyon) göre değişir:

#### a) Sınıflandırma Problemlerinde
- **Doğruluk (Accuracy):** Doğru tahminlerin oranı.
- **Kesinlik (Precision), Duyarlılık (Recall), F1 Skoru:** Özellikle dengesiz veri setlerinde önemlidir.
- **ROC-AUC:** Sınıflandırma eşiğinin farklı değerlerinde modelin başarı durumu.

#### b) Regresyon Problemlerinde
- **Ortalama Mutlak Hata (MAE), Ortalama Kare Hata (MSE), Kök Ortalama Kare Hata (RMSE):** Tahmin ve gerçek değerler arasındaki farkı ölçen skorlar.
- **R-Kare (R²):** Modelin değişkenliği ne kadar açıkladığını gösterir.

### 4. Overfitting ve Underfitting'in Tespiti
- Model hem eğitimde hem de testte yüksek başarı gösteriyorsa **ideal**dir.
- Model sadece eğitimde iyiyse, testte kötüyse **aşırı öğrenme (overfitting)** var demektir.
- Her iki sette de kötüyse **az öğrenme (underfitting)** olabilir.

**Özet:**  
Model başarı değerlendirme yöntemleri; modelin sadece eğitildiği değil, gerçek hayatta da iyi çalışıp çalışmadığını görmeyi sağlar. Doğru ayrım, uygun metrik ve testler ile güvenilir modeller geliştirilir.


## Model Doğrulama Yöntemleri Nedir?

Model doğrulama yöntemleri, makine öğrenmesi modellerinin oluşturulma sürecinde modelin ne kadar doğru ve güvenilir çalıştığını ölçmek için kullanılan yöntemlerdir.

Temel amaç, modelin sadece eğitim verilerinde değil, daha önce görmediği yeni verilere karşı da başarılı tahmin yapıp yapmadığını test etmektir. Eğitimde iyi sonuç veren bir modelin gerçek dünyadaki verilerde de başarılı olabilmesi için doğrulama gereklidir.

En yaygın model doğrulama yöntemleri:

- **Eğitim-Test Ayrımı (Train-Test Split):** Veri ikiye bölünür; bir kısmı eğitim, bir kısmı test için kullanılır.
- **Çapraz Doğrulama (Cross-Validation):** Veri birden fazla parçaya bölünür; her parça dönüşümlü olarak test seti olur, kalanlar eğitimde kullanılır.
- **Leave-One-Out:** Her örnek sırayla test için dışarıda bırakılır, kalanları eğitimde kullanılır (özellikle küçük veri setlerinde).

Doğrulama sayesinde modelin başarısı çeşitli metriklerle (doğruluk, hata oranı vb.) değerlendirilir. Ayrıca aşırı öğrenme (overfitting) veya az öğrenme (underfitting) olup olmadığı anlaşılır.



## Yanlılık-Varyans (Bias-Variance) Değiş Tokuşu (Trade-Off) Nedir?

Makine öğrenmesinde **yanlılık-varyans değiş tokuşu**, bir modelin tahminlerindeki iki temel hata kaynağı arasındaki dengeyi ifade eder:

- **Yanlılık (Bias):** Modelin, karmaşık ilişkileri yeterince öğrenememesi sonucu ortaya çıkan sistematik hatadır. Yani modelin "basitliğinden" kaynaklanan hata. Yüksek yanlılık, modelin hem eğitim hem de test verisinde kötü performans göstermesine sebep olur (**underfitting**).

- **Varyans (Variance):** Modelin, eğitim verisindeki rastlantısal dalgalanmalara aşırı uyum sağlamasıyla oluşan hatadır. Model, eğitim verisindeki neredeyse her ayrıntıyı "ezberler", ancak yeni (test) verilerde iyi genelleme yapamaz (**overfitting**).

**Değiş Tokuşun Temel Mantığı:**

- Çok basit modeller (ör: lineer regresyonla karmaşık bir ilişkiyi modellemek) genellikle yüksek **yanlılık** ve düşük **varyans** gösterir.
- Çok karmaşık modeller ise (ör: aşırı dallanmış bir karar ağacı gibi) düşük **yanlılık** ama yüksek **varyans** gösterebilir.

Amaç, ne çok yüksek yanlılığa (az öğrenen model) ne de çok yüksek varyansa (ezberleyen model) sahip olmak, yani model karmaşıklığını doğru ayarlayarak genelleme kabiliyetini artırmaktır.

**Görsel Açıklama:**  
Yüksek yanlılıkta oklar hedefin hep bir yanında toplanır (sistematik hata), yüksek varyansta oklar hedefin etrafına dağılmıştır (tahminler tutarsız). İdeal durumda oklar merkezin çevresinde ve yakın olur.

**Sonuç:**  
Başarılı bir makine öğrenmesi modeli geliştirmek için yanlılık ile varyans arasındaki dengeyi bulmak kritik öneme sahiptir. Bu denge, model seçimi, parametre ayarı ve doğrulama yöntemleriyle sağlanır.
