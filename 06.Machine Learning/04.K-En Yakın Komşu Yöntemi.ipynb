"""
# K En Yakın Komşu (K-Nearest Neighbors - KNN) Yöntemi Nedir?

K En Yakın Komşu (KNN) algoritması, makine öğrenmesinde kullanılan ve anlaşılması oldukça kolay olan bir yöntemdir. Hiç bilmeyen birisi için şöyle açıklayabiliriz:

---

## KNN Nasıl Çalışır?

1. **Varsayalım ki elimizde, özellikleri ve hangi gruba (sınıfa) ait olduğu bilinen bir sürü örnek (veri) var.**
2. Yeni gelen ve hangi sınıfa ait olduğunu bilmediğimiz bir örneği (örneğin yeni bir kişi veya nesne) sınıflandırmak istiyoruz.
3. Öncelikle, bu yeni örneğin mevcut örneklere (yani veri setimizdeki diğer verilere) olan benzerliğine bakarız. Buradaki benzerlik çoğu zaman iki örnek arasındaki mesafe olarak hesaplanır. (Bunu iki nokta arasındaki gerçek mesafe gibi düşünebiliriz.)
4. “K” sayıda, yani kullanıcının belirlediği sayı kadar (örneğin k=3, k=5 olabilir) en yakın örneği buluruz. Yani bu yeni örneğe en çok benzeyen K tane veri seçilir.
5. Bu K komşu örneğin hangi sınıfa (veya gruba) ait olduğuna bakılır. Kimin grubu çoğunluktaysa, yeni örneğimizin de o gruba ait olduğuna karar verilir.

---

## Bir Örnekle Açıklama

Diyelim ki elinizde kırmızı ve mavi toplardan oluşan bir grup var. Şimdi elinize rengi bilinmeyen yeni bir top geçti. Bu topun hangi gruba ait olduğunu bulmak istiyorsunuz. Elinizdeki toplar arasından yeni topa en yakın (örneğin rengine göre en benzer) 5 tane topu seçiyorsunuz. Eğer bu 5 toptan 3’ü kırmızı, 2’si mavi ise, yeni topun kırmızı olduğunu tahmin ediyorsunuz.

---

## Özetle

- KNN, yeni bir şeyin (örneğin bir kişinin hastalığı olup olmadığını) tahmin ederken, ona en çok benzeyen örneklere bakar.
- Komşuları hangi gruptaysa, yeni olanı da o gruba yerleştirir.
- Herhangi bir karmaşık matematik ya da model eğitimi gerekmez, oldukça sezgiseldir ve kolayca uygulanabilir.

"""

```python
import pandas as pd
from sklearn.metrics import classification_report, roc_auc_score
from sklearn.model_selection import GridSearchCV, cross_validate
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 500)
```

## 1. Exploratory Data Analysis

```python
df = pd.read_csv("datasets/diabetes.csv")

df.head()
```

```python
df.shape
```

```python
df.describe().T
```

```python
df["Outcome"].value_counts()
```

### 2. Data Preprocessing & Feature Engineering


```python
# Bağımlı (hedef) ve bağımsız değişkenlerin (özelliklerin) ayrılması
y = df["Outcome"]
X = df.drop(["Outcome"], axis=1)

# Özelliklerin standartlaştırılması (standart skala ile)
X_scaled = StandardScaler().fit_transform(X)

# Standartlaştırılmış veriler DataFrame'e dönüştürülür
X = pd.DataFrame(X_scaled, columns=X.columns)
```

#### 3. Modeling & Prediction

```python
# KNN modeli oluşturulup eğitilirken bazı önemli parametreler bulunmaktadır:
# n_neighbors: Komşu sayısı (varsayılan 5)
# metric: Mesafe ölçütü, varsayılan olarak "minkowski" kullanılır (genellikle öklidyen mesafe ile aynıdır).
# weights: Komşuların katkı ağırlığı ("uniform" = eşit ağırlık, "distance" = uzaklığa göre ağırlıklandırma).
# Burada varsayılan parametreler ile model kuruldu:

knn_model = KNeighborsClassifier(
    n_neighbors=5,         # Komşu sayısı
    metric="minkowski",    # Mesafe ölçütü
    weights="uniform"      # Komşu ağırlığı
).fit(X, y)

# Rastgele bir kullanıcı seçilir
random_user = X.sample(1, random_state=45)

# Rastgele kullanıcının sınıfı tahmin edilir
knn_model.predict(random_user)
```

### 4. Model Evaluation


```python
# Modelin doğruluğunu ve çeşitli metriklerle başarımını ölçelim:

# Tüm veride tahminler (y_pred) ve olasılıklar (y_prob)
y_pred = knn_model.predict(X)
y_prob = knn_model.predict_proba(X)[:, 1]

# Sınıflandırma raporu: accuracy, precision, recall, f1-score gibi birçok metrik içerir
print("Sınıflandırma Raporu:\n", classification_report(y, y_pred))

# AUC (Area Under Curve) metriği, modelin ayrım gücünü gösterir (1'e ne kadar yakınsa o kadar iyi)
auc_score = roc_auc_score(y, y_prob)
print("AUC Skoru:", auc_score)

# 5 katlı çapraz doğrulama ile modelin genel başarısı
cv_results = cross_validate(knn_model, X, y, cv=5, scoring=["accuracy", "f1", "roc_auc"])

print("Çapraz Doğrulama Sonuçları (Ortalama):")
print("Accuracy:", cv_results['test_accuracy'].mean())
print("F1 Score:", cv_results['test_f1'].mean())
print("ROC AUC:", cv_results['test_roc_auc'].mean())

# Bazı ideal skorlar örnek olması açısından (değişebilir)
# acc ~ 0.73
# f1  ~ 0.59
# auc ~ 0.78

# Modelin parametrelerine yakından bakalım:
print("\nKNN Modelinin Parametreleri:")
params = knn_model.get_params()
for param, value in params.items():
    print(f"{param}: {value}")

# Temel parametreler:
# - n_neighbors: Kaç komşudan oy alınacağı
# - metric: Mesafe ölçütü (örn. 'minkowski', öklidyen için p=2)
# - weights: Komşu oylamasında ağırlık tipi ('uniform', 'distance')
# - algorithm: Komşuluk arama algoritması (otomatik seçilir veya elle 'ball_tree', 'kd_tree', 'brute')
# - leaf_size: BallTree/KDTree için yaprak düğüm boyutu, hız/performans için önemli
# - p: Minkowski metriğinde derece (p=1 Manhattan, p=2 Öklidyen)
# - n_jobs: Paralel işlemci sayısı
# ve daha fazlası...

# Model başarısını artırmak için:
# 1. Veri boyutu ve çeşitliliği artırılabilir.
# 2. Daha iyi veri ön işleme ve eksik değer işlemleri yapılabilir.
# 3. Özellik seçimi/mühendisliğiyle önemli değişkenler vurgulanabilir.
# 4. KNN'nin n_neighbors, metric ve weights gibi hiperparametreleri GridSearch/Cross Validation ile optimize edilebilir.
```

#### 5. Hyperparameter Optimization

```python
# Burada hiperparametre optimizasyonu için GridSearchCV ile KNN algoritmasının en iyi parametrelerini buluyoruz.
# Öncelikle bir KNeighborsClassifier modeli oluşturuluyor ve varsayılan parametreleri aşağıdaki şekilde elde edebilirsiniz:
print("KNN Model Varsayılan Parametreleri:\n")
for param, value in KNeighborsClassifier().get_params().items():
    print(f"{param}: {value}")

# optimize edilecek parametreleri tanımlıyoruz. Bu örnekte sadece 'n_neighbors' (kaç komşu bakılacağı) için 2'den 49'a kadar olan değerler denenecek.
# Diğer önemli parametreler de optimize edilebilir (ör: 'metric', 'weights', 'algorithm' gibi), ancak burada sadece en temel olanı seçtik.
knn_params = {"n_neighbors": range(2, 50)}

# GridSearchCV ile 5 katlı çapraz doğrulama yapıyoruz. n_jobs=-1 ile tüm işlemciler kullanılır, verbose=1 ile süreç ekrana yazdırılır.
knn_model = KNeighborsClassifier()

knn_gs_best = GridSearchCV(
    knn_model,
    knn_params,
    cv=5,
    n_jobs=-1,
    verbose=1
).fit(X, y)

print("GridSearch sonrası en iyi parametreler:", knn_gs_best.best_params_)

# KNN'de başlıca parametreler şunlardır:
# - n_neighbors: Kaç komşuya bakılacağı, model karmaşıklığı ve genelleme performansını ciddi şekilde etkiler.
# - metric: Mesafe ölçütü, varsayılanı 'minkowski' (p=2 için öklidyen, p=1 için manhattan).
# - weights: Komşuluk oyu ağırlıkları; 'uniform' (eşit ağırlık) ya da 'distance' (yakın olanlar daha fazla oy verir).
# - algorithm: Komşu arama algoritması; 'auto', 'ball_tree', 'kd_tree', 'brute'.
# - leaf_size: ağaç-temelli algoritmalarda performans ayarı için.
# - p: 'minkowski' metriği için derece, p=2 öklidyen, p=1 manhattan.
# - n_jobs: Hesaplama için kullanılacak CPU çekirdeği sayısı.
```

### Final Model

```python
# Final model, GridSearchCV sonucunda bulunan en iyi parametrelerle eğitilecek ve çapraz doğrulama ile nihai performansı ölçülecek.

# GridSearchCV sonucunda elde edilen en iyi hiperparametrelerle KNN modelini tekrar kuruyoruz.
knn_final = knn_model.set_params(**knn_gs_best.best_params_).fit(X, y)

# Modelin nihai başarısını değerlendirmek için tekrar cross_validate ile doğrulama yapıyoruz.
cv_results = cross_validate(
    knn_final,
    X,
    y,
    cv=5,
    scoring=["accuracy", "f1", "roc_auc"]
)

print(f"Ortalama Accuracy: {cv_results['test_accuracy'].mean():.2f}")
print(f"Ortalama F1 Score: {cv_results['test_f1'].mean():.2f}")
print(f"Ortalama ROC AUC: {cv_results['test_roc_auc'].mean():.2f}")

# Model ile yeni (rasgele seçilmiş) bir kullanıcının çıktısını (tahminini) test edelim.
random_user = X.sample(1)
prediction = knn_final.predict(random_user)
print("Rasgele seçilen kullanıcının tahmini:", prediction)

# --- Açıklama ---
# Bu bölümde kullanılan KNeighborsClassifier parametreleri:
# - n_neighbors: Komşu sayısı (GridSearchCV sonucu en iyi değer)
# - metric: Mesafe hesaplama yöntemi (varsayılan: 'minkowski')
#   - 'minkowski' için p=2, yani Öklidyen mesafe; p=1 ise Manhattan.
# - weights: Oy ağırlığı (varsayılan: 'uniform', yani tüm komşular eşit oy ağırlıklı)
# - algorithm: Komşu arama algoritması (varsayılan: 'auto')
# - leaf_size: Ağaç tabanlı arama algoritmaları için (default: 30)
# - p: Minkowski metriği için derece (default: 2)
# - n_jobs: Hesaplama için kullanılacak işlemci sayısı (default: None; cross_validate ve GridSearchCV'de n_jobs=-1 çoklu çekirdek kullanır)
```
