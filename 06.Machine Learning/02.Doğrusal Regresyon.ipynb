# 📊 **Doğrusal Regresyonu Sıfırdan Anlama**


---

### 📎 Amaç:

Bir şeyin (örn. not, maaş, ev fiyatı) başka değişkenlerle (örn. yaş, tecrübe, oda sayısı) **nasıl ilişkili olduğunu** en temel matematik kurallarıyla ve sayısal olarak tahmin etmeye çalışmak.

---

### 🧩 **En Temel Formül**

En basit doğrusal regresyon şu şekilde yazılır:

$$
\hat{y}_i = b + w x_i
$$

**Açıklamalar:**
- $\hat{y}_i$ : Tahmin edilen değer (ör. ev fiyatı, maaş)
- $b$ : Sabit terim (tüm değişkenler sıfırsa başlangıç değeri)
- $w$ : Katsayı, "bir birim değişim sonucu sonuç ne kadar değişir?" bilgisini verir
- $x_i$ : Bağımsız değişkenin değeri (ör. oda sayısı, yaş)

---

### 🧮 **Birden Fazla Değişken Durumu**

Birden fazla faktör birlikte etkiliyorsa (ör. hem yaş hem eğitim yılı gibi), denklem şöyle olur:

$$
\hat{y} = b + w_1 x_1 + w_2 x_2 + w_3 x_3 + \dots + w_p x_p
$$

**Yani:**
- Her değişken ($x_1, x_2, ...$) kendi katsayısıyla ($w_1, w_2, ...$) çarpılır.
- Toplamları, sabit terim $b$'ye eklenir.
- Sonuç: Modele göre tahmin edilen değer.

---

### 📝 **Özetle:**

- Değişkenlerle sonuç (örn. oda sayısı → ev fiyatı) arasında **düz bir çizgi çekmek** gibi düşünülebilir.
- Zor matematik gerekmez: çarp, topla!
- Amaç: “Bir faktör bir birim değişirse, sonuç ne kadar değişir?” sorusunu sayısal olarak görebilmek.

---

Doğrusal regresyonda amaç, modelin tahminleriyle gerçek değerler arasındaki farkların karelerinin toplamını (veya ortalamasını) en küçük yapmaktır. Bunu başarmak için b ve w katsayıları en uygun şekilde seçilir. Yani, model veriye öyle bir düz çizgi çizer ki, tüm noktalar ile çizgi arasındaki dikey mesafelerin karelerinin toplamı minimum olur.

Kısaca, modelimiz gerçek ve tahmin edilen değerler arasındaki hata miktarını minimuma indirir. Hesaplanan hata genellikle ortalama karesel hata (MSE) olarak adlandırılır ve model bu değeri küçültmeye çalışır.

En temel formül:
$$
\hat{y}_i = b + w x_i
$$

Buradaki amaç, çizgiyi toplam tahmin hatası en az olacak şekilde belirlemektir.

### 🎯 Tahmin Hatası Nasıl Ölçülür? (MSE, RMSE, MAE)

Bir makine öğrenmesi modelinin tahminlerinin ne kadar doğru olduğunu ölçmek için bazı temel hata metrikleri kullanılır. Bunlar sayesinde modelimizin tahminlerinin gerçek değerlere ne kadar yakın olduğunu sayısal olarak görebiliriz. En yaygın üç hata ölçütü şunlardır:

---

#### 1. MSE (Mean Squared Error) — Ortalama Kare Hata

MSE, her bir tahminin gerçek değerden ne kadar saptığını ölçer.

- Önce tahmin ile gerçek değer arasındaki fark (hata) bulunur.
- Bu farkın karesi alınır (eksi işaretleri ve büyük hatalar daha çok dikkate alınsın diye).
- Tüm kareler toplanır ve gözlem sayısına bölünür.

Formülü:
$$
MSE = \frac{1}{n} \sum_{i=1}^n (y_i - \hat{y}_i)^2
$$

Burada:
- $y_i$: $i$. gözlemdeki gerçek değer
- $\hat{y}_i$: $i$. gözlemdeki model tahmini
- $n$: Toplam gözlem (örnek) sayısı

---

#### 2. RMSE (Root Mean Squared Error) — Kök Ortalama Kare Hata

RMSE, MSE’nin karekökünü alır. Böylece hata, tahmin edilen değişkenin birimiyle aynı biçimde ifade edilir.

Formülü:
$$
RMSE = \sqrt{ \frac{1}{n} \sum_{i=1}^n (y_i - \hat{y}_i)^2 }
$$

- RMSE’yi yorumlamak MSE’den daha kolaydır; çünkü örneğin fiyat tahmini yapıyorsanız biriminiz yine para cinsindendir.

---

#### 3. MAE (Mean Absolute Error) — Ortalama Mutlak Hata

MAE, gerçek ve tahmin değerleri arasındaki farkların sadece büyüklüğüne bakar.

- Hata farklarının mutlak değeri alınır (yani eksilere bakılmaz).
- Toplanır ve ortalaması alınır.

Formülü:
$$
MAE = \frac{1}{n} \sum_{i=1}^n |y_i - \hat{y}_i|
$$

---

#### Özet

- **MSE** ve **RMSE**, büyük hataları daha çok önemsediğinden uç (çok büyük) hatalara duyarlıdır.
- **MAE** ise tüm hataları eşit şekilde değerlendirir ve uç değerlere karşı daha dayanıklıdır.
- Bu metrikler modelimizin tahminlerinin kalitesini sayısal olarak değerlendirmemize yardımcı olur.

### Parametrelerin Tahmin Edilmesi (Ağırlıkların Bulunması)

Makine öğrenmesinde doğrusal regresyon modelinin kuralları belirlendikten sonra, modeli oluşturan katsayıların (parametrelerin, ağırlıkların) nasıl bulunacağını anlamamız gerekir. Bu işlem, modelin **öğrendiği** kısmı oluşturur.

---

#### **Analitik Çözüm: Normal Denklemler Yöntemi (En Küçük Kareler Yöntemi)**

Bu yöntemde, bir doğrusal regresyon modelindeki parametreler için doğrudan bir çözüm formülü vardır ve bu formüller kullanılarak ağırlıklar hesaplanabilir. Burada en önemli amaç, modelin tahminleri ile gerçek değerler arasındaki farkların karelerinin toplamını (SSE - Sum of Squared Errors) **minimum** yapmak yani hata miktarını en aza indirmektir.

---

#### **Temel Kavramlar ve Formüller**

Diyelim ki bir veri kümenizde $\mathbf{n}$ tane $(x_i, y_i)$ veri çifti var. Burada:

- $x_i$: $i$. gözlem noktasındaki giriş (bağımsız değişken/değer)
- $y_i$: $i$. gözlemdeki gerçek çıktı (bağımlı değişken/değer)
- $\hat{y}_i$: $i$. gözlemde modelin tahmini

Doğrusal regresyonun temel (basit-tek değişkenli) formülü:

$$
\hat{y}_i = b_0 + b_1 x_i
$$

Buradaki $b_0$ **(bias/ya da sabit terim/intercept)**, $b_1$ ise **(eğim/weight katsayısı)**dır.

---

#### **Tahmin Hatası ve Optimizasyon**

Modelimizin tahminlerinin, gerçek değerlerden sapmasını kareleriyle topladığımızda oluşan değere **SSE (Toplam Kare Hata)** denir:

$$
SSE = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \sum_{i=1}^n [y_i - (b_0 + b_1 x_i)]^2
$$

veya çoklu doğrusal regresyon için daha genel olarak:

$$
SSE = \sum_{i=1}^n (y_i - \hat{y}_i)^2
$$

Bizim amacımız, parametreleri bu SSE'yi **en küçük** yapacak şekilde seçmektir.

---

#### **Parametrelerin Analitik Olarak Hesaplanması**

##### Tek Değişkenli (Basit) Doğrusal Regresyonda:

Kapalı formüller:

- $b_1$ (eğim/weight katsayısı):

$$
b_1 = \frac{\sum_{i=1}^{n} (x_i - \bar{x})\,(y_i - \bar{y})}{\sum_{i=1}^{n} (x_i - \bar{x})^2}
$$

Burada $\bar{x}$ tüm $x$ değerlerinin ortalaması; $\bar{y}$ ise tüm $y$ değerlerinin ortalamasıdır.

- $b_0$ (bias/intercept):

$$
b_0 = \bar{y} - b_1 \bar{x}
$$

##### Çoklu Doğrusal Regresyon (Multiple Linear Regression):

Birden fazla bağımsız değişken ($p$ tane) olduğunda model ve parametre tahmini matrissel olarak yazılır. $\mathbf{X}$ giriş değerleri matrisi (satırlar gözlemler, sütunlar değişkenler), $\mathbf{Y}$ ise gerçek değer vektörüdür.

Model:
$$
\mathbf{\hat{y}} = \mathbf{X} \boldsymbol{\beta}
$$

Analitik (kapalı) çözüm:
$$
\boldsymbol{\hat{\beta}} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{Y}
$$

- $\boldsymbol{\hat{\beta}}$: Parametreler (ağırlıklar) vektörü.  
- $\mathbf{X}^T$: $\mathbf{X}$'in transpozu  
- $(\mathbf{X}^T \mathbf{X})^{-1}$: Bu matrisin tersi alınır (tersi alınabiliyorsa)

Bu formül, aynı anda birden fazla (çoklu) açıklayıcı değişken olduğunda parametreleri doğrudan hesaplamak için kullanılır.

---

#### **Alternatif: Optimizasyon Çözümü – Gradyan İnişi (Gradient Descent)**

Büyük veri setlerinde veya çok fazla açıklayıcı değişken olduğunda, analitik çözümün uygulanması pratik olmayabilir veya çok pahalıya mal olabilir. Bu durumlarda **Gradyan İnişi (Gradient Descent)** gibi optimizasyon algoritmaları kullanılır.

Gradient descent yaklaşımında parametreler, hata fonksiyonunun gradyanını takip ederek adım adım güncellenir ve minimum hata noktasına ulaşılır.

Güncellenen parametreler aşağıdaki formüllerle elde edilir:

$$
\theta_0 := \theta_0 - \alpha \frac{1}{m} \sum_{i=1}^m \Big(h_\theta (x^{(i)}) - y^{(i)}\Big)
$$

$$
\theta_1 := \theta_1 - \alpha \frac{1}{m} \sum_{i=1}^m \Big(h_\theta (x^{(i)}) - y^{(i)}\Big)x^{(i)}
$$

Burada:
- $\theta_0$, $\theta_1$: Öğrenilen parametreler
- $\alpha$: Öğrenme oranı (learning rate)
- $m$: Gözlem sayısı
- $h_\theta(x^{(i)})$: $i$. gözlemde modelin tahmini


---

#### **Özetle**

- Doğrusal regresyonun amacı, veriye en iyi uyan doğruyu veya düzlemi bulmaktır.
- Tek değişkenli durumda ($b_0$, $b_1$), çok değişkenli durumda ($\boldsymbol{\beta}$ vektörü) yukarıdaki formüllerle tüm parametreler analitik olarak hesaplanabilir.
- Analitik çözüm uygulanamaz hale geldiğinde, gradyan inişi gibi optimizasyon algoritmaları ile parametreler adım adım öğrenilir.
- **Sonuç:**  
  Elimizdeki gözlemlere dayanarak, doğrusal regresyonun ağırlıklarını (parametrelerini) ister analitik (kapalı formül), ister optimizasyon algoritması ile matematiksel olarak belirlemiş oluruz.

### **Gradyan İnişi (Gradient Descent) Nedir?**

Gradyan inişi, makine öğrenmesinde ve istatistikte kullanılan bir **optimizasyon algoritmasıdır**. En genel anlamıyla, bir fonksiyonun minimum (veya bazen maksimum) değerini bulmak için kullanılır. Doğrusal regresyon gibi modellerde ise amacımız parametreleri, tahmin edilen değerlerle gerçek değerler arasındaki hatayı (kayıp fonksiyonunu) en aza indirecek şekilde ayarlamaktır.

#### Gradyan İnişinin Temel Adımları

1. Başlangıçta parametreler (ağırlıklar ve bias gibi) rastgele veya bir değerle başlatılır.
2. Kayıp fonksiyonu bu parametrelerle hesaplanır.
3. Kayıp fonksiyonunun parametrelere göre gradyanı (türevi) bulunur. Gradyan, fonksiyonun hangi yönde daha hızlı azalacağını gösterir.
4. Parametreler, gradyanın ters yönünde ve bir “öğrenme oranı (learning rate)” kadar güncellenir.
5. Bu adımlar, kayıp fonksiyonu yeterince küçük olana kadar veya belirli bir iterasyon sayısına kadar tekrarlanır.

#### Matematiksel Gösterim

$$
\theta := \theta - \alpha \cdot \nabla_\theta J(\theta)
$$

- $\theta$: Modelin parametreleri
- $\alpha$: Öğrenme oranı (adım büyüklüğü)
- $J(\theta)$: Kayıp fonksiyonu
- $\nabla_\theta J(\theta)$: Kayıp fonksiyonunun parametreler üzerindeki gradyanı

#### Avantajları
- Büyük veri setlerinde analitik çözüme göre daha hızlı ve pratiktir.
- Nümerik yaklaşımı sayesinde karmaşık modellerde de kullanılabilir.

#### Dezavantajları
- Uygun öğrenme oranı seçilmezse, minimuma ulaşamayabilir veya yavaş ilerleyebilir.
- Yerel minimumlarda takılabilir.

**Özet:**  
Gradyan inişi, iteratif olarak model parametrelerini kayıp fonksiyonunu azaltacak şekilde güncelleyerek en iyi modelin elde edilmesini sağlar.

# Sales Prediction with Linear Regression


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.float_format', lambda x: '%.2f' % x)

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split, cross_val_score
```

```python
# Simple Linear Regression with OLS Using Scikit-Learn

df = pd.read_csv("datasets/advertising.csv")
df.shape
```

```python
# 1. Bağımsız değişken (X) ve bağımlı değişkeni (y) seçiyoruz.
X = df[["TV"]]        # Televizyon reklam harcaması bağımsız değişken olarak alındı.
y = df[["sales"]]     # Satışlar bağımlı değişken olarak alındı.

# 2. Lineer Regresyon modelimizi oluşturup, verimize fit (uydurma) ediyoruz.
reg_model = LinearRegression()       # Model nesnesi oluşturuluyor.
reg_model.fit(X, y)                  # Model, X ve y verilerine göre öğreniyor.

# 3. Modelin sabit (intercept/beta0) katsayısını inceliyoruz.
print("Modelin sabit terimi (intercept):", reg_model.intercept_[0])
```

```python
# Diyelim ki TV reklamına 150 birim para harcadık. Acaba bu durumda ne kadar satış olmasını bekleriz?
# Bunu anlamak için modelimizin kurduğu matematiksel ilişkiyi kullanıyoruz:
# Beklenen satış = Sabit terim (intercept) + TV harcaması katsayısı (coef) * 150
# Yani;
beklenen_satis = reg_model.intercept_[0] + reg_model.coef_[0][0]*150
print("150 birimlik TV harcaması için beklenen satış miktarı:", beklenen_satis)
```

```python
# Şimdi de, TV reklamına 500 birim para harcandığında beklenen satış miktarını adım adım hesaplayalım.

# 1. Adım: Modelimizdeki sabit terimi (intercept) buluyoruz.
sabit_terim = reg_model.intercept_[0]
print(f"Modelin sabit terimi (intercept): {sabit_terim}")

# 2. Adım: TV değişkeninin katsayısını (coef) buluyoruz.
tv_katsayisi = reg_model.coef_[0][0]
print(f"TV harcaması katsayısı (coef): {tv_katsayisi}")

# 3. Adım: TV harcaması için 500 birim değerini modele yerleştiriyoruz.
tv_harcamasi = 500

# 4. Adım: Beklenen satış miktarını formüle göre hesaplıyoruz:
# Beklenen satış = sabit_terim + tv_katsayisi * tv_harcamasi
beklenen_satis_500 = sabit_terim + tv_katsayisi * tv_harcamasi

print(f"500 birimlik TV harcaması için beklenen satış miktarı: {beklenen_satis_500}")
```

```python
df.describe().T
```

```python
# Modelin Görselleştirilmesi
# Burada doğrusal regresyon sonucu elde ettiğimiz model denklemini ve terimleri anlamlandırıyoruz.
# Model denklemi şu şekildedir: Sales = b0 + b1*TV
#   - b0: Sabit terim (intercept), TV harcaması 0 olduğunda beklenen satış miktarıdır.
#   - b1: TV harcamasının katsayısı (coef), TV harcamasındaki bir birimlik artışın satışa etkisini gösterir.
#   - TV: Bağımsız değişkenimiz, yani reklam harcaması.
#   - Sales: Bağımlı değişkenimiz, yani satış miktarı.

g = sns.regplot(x=X, y=y, scatter_kws={'color': 'b', 's': 9},
                ci=False, color="r")

g.set_title(
    f"Model Denklemi: Sales = {round(reg_model.intercept_[0], 2)}"
    f"+ TV*{round(reg_model.coef_[0][0], 2)}"
)
g.set_ylabel("Satış Sayısı (Sales)")
g.set_xlabel("TV Harcamaları (TV)")
plt.xlim(-10, 310)
plt.ylim(bottom=0)
plt.show()
```

#### Tahmin Başarısı

```python
# MSE (Mean Squared Error - Ortalama Kare Hata) modelimizin tahminlerinin doğruluğunu ölçen önemli bir metriktir.
# Gerçek satış değerleri (y) ile modelimizin tahmin ettiği satış değerleri (y_pred) arasındaki farkların karelerinin ortalamasını hesaplarız.
y_pred = reg_model.predict(X)  # Modelimiz bütün X (TV harcamaları) için tahmin değerleri üretir.
mse = mean_squared_error(y, y_pred)  # MSE hesaplanır: düşük olması modelin daha iyi olduğunu gösterir.
print(f"Modelin Ortalama Kare Hatası (MSE): {mse}")
```

```python
# RMSE (Root Mean Squared Error - Kök Ortalama Kare Hata) modeli değerlendirmek için kullanılan önemli bir metriktir.
# RMSE, modelin tahmin ettiği değerlerle gerçek değerler arasındaki farkların karesinin ortalamasının kareköküdür.
# Yani, modelimizin tahminlerde ortalama olarak ne kadar hata yaptığını yorumlamamızı sağlar ve hata biriminin orijinal değişkenimizle (sales) aynı olmasını sağlar.
rmse = np.sqrt(mean_squared_error(y, y_pred))
print(f"Modelin Kök Ortalama Kare Hatası (RMSE): {rmse}")
# Örneğin, 3.24 çıktısı modelimizin tahminlerinin ortalama 3.24 civarında hata yaptığını gösterir.
```

```python
# MAE (Mean Absolute Error - Ortalama Mutlak Hata), modelimizin tahmin ettiği değerlerle gerçek değerler arasındaki farkların mutlak değerlerinin ortalamasıdır.
# Yani, modelimizin her bir tahmininde ortalama olarak ne kadar hata yaptığını gösterir. 
# Örneğin, MAE değeri 2.5 ise, modelimiz ortalama olarak her tahmininde gerçek değerden 2.5 birim şaşmaktadır.
mae = mean_absolute_error(y, y_pred)
print(f"Modelin Ortalama Mutlak Hatası (MAE): {mae}")
```

```python
# R-KARE veya Determinasyon Katsayısı (R^2), modelin bağımlı değişkendeki (sales) toplam varyansın ne kadarını açıkladığını gösterir.
# 1'e yaklaştıkça modelin başarı oranı artar. 0'a yakınsa model başarısız demektir.
# scikit-learn'de .score() fonksiyonu aslında R-kare değerini döndürür.
r2 = reg_model.score(X, y)
print(f"Modelin Determinasyon Katsayısı (R^2): {r2}")
```

### Multiple Linear Regresyon

```python
# Öncelikle "datasets/advertising.csv" adlı veri setimizi pandas kütüphanesi ile okuyoruz ve bir DataFrame'e yüklüyoruz.
df = pd.read_csv("datasets/advertising.csv")

# Ardından, veri setimizdeki bağımsız değişkenleri (features) belirliyoruz. Burada 'sales' sütununu çıkararak geriye tüm bağımsız değişkenler (TV, radio, newspaper) kalıyor.
X = df.drop('sales', axis=1)

# Son olarak, tahmin etmek istediğimiz bağımlı değişkenimizi yani hedef değişkenimizi seçiyoruz. Burada 'sales' sütunu hedef değişken olarak belirleniyor ve çift köşeli parantez ile DataFrame olarak alınmış.
y = df[["sales"]]
```

```python
# Modelimizi eğitmek ve test etmek için veri setimizi ikiye böleceğiz: eğitim ve test seti.
# train_test_split fonksiyonu, veri setini belirtilen oranlarda eğitim ve test olarak böler.
# test_size=0.20 parametresi, verinin %20'sinin test setine, %80'inin ise eğitim setine ayrılmasını sağlar.
# random_state=1 ise, her çalıştırmada aynı şekilde bölünmeyi garanti eder; böylece sonuçlar tekrarlanabilir olur.
# X : Bağımsız değişkenler (features), 
# y : Bağımlı değişken (hedef/target)
# train_test_split fonksiyonu sırasıyla; eğitime gidecek bağımsız değişkenler (X_train), testte kullanılacak bağımsız değişkenler (X_test),
# eğitime gidecek bağımlı değişkenler (y_train), testte kullanılacak bağımlı değişkenler (y_test) olarak dört değer döndürür.
X_train, X_test, y_train, y_test = train_test_split(
    X,             # Bağımsız değişkenler (özellikler/features)
    y,             # Bağımlı değişken (hedef/target)
    test_size=0.20,  # Test setinin oranı (%20)
    random_state=1   # Rastgelelik için sabit değer, yeniden çalıştırıldığında aynı şekilde böler
)
```

```python
# Basit Doğrusal Regresyonu Gradient Descent ile yaparken kullandığımız cost_function(Y, b, w, X) fonksiyonunu inceleyelim ve uygulayalım.
# Bu fonksiyon, verilen b (bias) ve w (ağırlık) parametreleri için Mean Squared Error (Ortalama Kare Hata) değerini hesaplar.
# Amacımız, b ve w değerlerini öyle ayarlamak ki cost_function sonucu (MSE) minimum olsun.


def cost_function(Y, b, w, X):
    """
    Mean Squared Error (Ortalama Kare Hata) hesaplayan maliyet fonksiyonu.

    Parametreler:
    -----------
    Y : array-like
        Gerçek (hedef) değerleri içeren liste ya da dizidir.
    b : float
        Bias (sabit) terimi, yani modelin öngörülen sabit başlangıç noktası.
    w : float
        Ağırlık katsayısı, girdiye (X) uygulanan çarpan.
    X : array-like
        Bağımsız değişkenler (özellik/girdi) listesi ya da dizisi.
    
    Fonksiyonun Adım Adım Açıklaması:
    -------------------------------
    1. m = len(Y): Veri setindeki gözlem (örnek) sayısı bulunur.
    2. sse = 0: Toplam karesel hatayı biriktirmek için bir değişken başlatılır.
    3. for i in range(m): Her bir veri noktası için döngüye girilir.
        a. y_hat = b + w * X[i]: O gözlem için modelin tahmini hesaplanır.
        b. y = Y[i]: Gerçek değer alınır.
        c. sse += (y_hat - y) ** 2: Tahmin ve gerçek değer arasındaki farkın karesi sse'ye eklenir.
    4. mse = sse / m: Toplam hata gözlem sayısına bölünerek ortalama kare hata elde edilir.
    5. return mse: Fonksiyon, hesaplanan MSE değerini döndürür.
    """
    m = len(Y)  # Gözlem sayısı
    sse = 0     # Toplam karesel hata başlatılır
    for i in range(m):  # Her gözlem için
        y_hat = b + w * X[i]  # Tahmini değer hesaplanır
        y = Y[i]              # Gerçek değer alınır
        sse += (y_hat - y) ** 2  # Hata karesi eklenir
    mse = sse / m  # Ortalama alınır
    return mse     # Maliyet döndürülür
```

```python
# update_weights fonksiyonu, verilen b ve w başlangıç değerlerini ve öğrenme oranını (learning_rate) kullanarak gradyan (türetilmiş maliyet fonksiyonu) yöntemine göre yeni ağırlıkları ve bias'ı (b, w) günceller.

def update_weights(Y, b, w, X, learning_rate):
    """
    Gradient Descent adımıyla b (bias) ve w (ağırlık) değerlerini günceller.

    Parametreler:
    -------------
    Y : array-like
        Gerçek (hedef) değerleri içeren dizi/liste.
    b : float
        Mevcut bias (sabit) terimi.
    w : float
        Mevcut ağırlık (katsayı) değeri.
    X : array-like
        Bağımsız değişken değerleri.
    learning_rate : float
        Öğrenme oranı (adım büyüklüğü).

    Fonksiyonun Adım Adım Açıklaması:
    ------------------------------------
    1. Her örnek için y_hat (tahmin) hesaplanır: y_hat = b + w * X[i]
    2. b terimi için türevler (hataların toplamı): (y_hat - y)
    3. w terimi için türevler (girdiyle çarpılmış hata toplamı): (y_hat - y) * X[i]
    4. Türevlerin ortalaması alınır ve öğrenme oranı ile çarpılarak mevcut değerlerden çıkarılır
    5. Güncellenmiş b ve w yeni değerler olarak döndürülür

    Returns:
    ---------
    new_b : float
        Güncellenmiş bias değeri.
    new_w : float
        Güncellenmiş ağırlık değeri.
    """
    m = len(Y)
    b_deriv_sum = 0
    w_deriv_sum = 0
    for i in range(m):
        y_hat = b + w * X[i]
        y = Y[i]
        b_deriv_sum += (y_hat - y)
        w_deriv_sum += (y_hat - y) * X[i]
    new_b = b - (learning_rate * 1 / m * b_deriv_sum)
    new_w = w - (learning_rate * 1 / m * w_deriv_sum)
    return new_b, new_w
```

```python
def train(Y, initial_b, initial_w, X, learning_rate, num_iters):
    """
    Gradient Descent algoritmasının eğitim (fit) fonksiyonudur.
    
    Adım adım:
    1. Başlangıçta bias (b) ve ağırlık (w) ilk değerleriyle başlar.
    2. Her iterasyonda:
        - Ağırlıklar güncellenir (update_weights fonksiyonu ile).
        - Güncel ağırlıklarla maliyet (cost_function - genelde MSE) hesaplanır.
        - Her adımda maliyet (hata) cost_history listesine eklenir.
        - Belirli aralıklarda (ör. i her 100 olduğunda), anlık b, w ve mse (hata) yazdırılır.
    3. Son durumda: 
        - Tüm iterasyon sonrası bulunan en iyi ağırlıklar, bias ve maliyet döndürülür.

    Parametreler:
    -------------
    Y : array-like
        Hedef değerler (bağımlı değişken).
    initial_b : float
        Başlangıç bias (sabit terim) değeri.
    initial_w : float
        Başlangıç ağırlık (katsayı) değeri.
    X : array-like
        Girdi verisi (bağımsız değişken).
    learning_rate : float
        Öğrenme oranı (adım büyüklüğü).
    num_iters : int
        İterasyon (dönüş) sayısı.

    Returns:
    ---------
    cost_history : list
        Tüm iterasyonlarda elde edilen hata (mse) değerleri.
    b : float
        Son bias değeri.
    w : float
        Son ağırlık değeri.
    """
    print("Başlangıç: b = {}, w = {}, mse = {}".format(initial_b, initial_w, cost_function(Y, initial_b, initial_w, X)))
    b = initial_b
    w = initial_w
    cost_history = []

    for i in range(num_iters):
        b, w = update_weights(Y, b, w, X, learning_rate)
        mse = cost_function(Y, b, w, X)
        cost_history.append(mse)
        if i % 100 == 0:
            print("iter = {} | b = {:.2f} | w = {:.4f} | mse = {:.4f}".format(i, b, w, mse))

    print("Sonuç: iterasyon = {}, b = {:.2f}, w = {:.4f}, mse = {:.4f}".format(num_iters, b, w, cost_function(Y, b, w, X)))
    return cost_history, b, w
```
