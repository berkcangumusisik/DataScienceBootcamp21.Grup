### Temel SQL Komutları 

SQL, veritabanlarında veri eklemek, okumak, değiştirmek ve silmek için kullanılan bir dildir. Temel konuları aşağıda bulabilirsin.

---

#### 1. Veriler Üzerinde İşlem Yapmak (DML)

Tablolar üzerindeki verilerle çalışmak için kullanılır.

**SELECT**  
Tablodaki verileri görüntüler. Tüm verileri görmek için:

```sql
SELECT * FROM ogrenci;
```

Belirli sütunları seçmek için:

```sql
SELECT ad, soyad FROM ogrenci;
```

Şart eklemek için:

```sql
SELECT * FROM ogrenci WHERE yas > 18;
```

---

**INSERT**  
Tabloya yeni satır (kayıt) ekler.

```sql
INSERT INTO ogrenci (ad, soyad, yas) VALUES ('Ali', 'Vural', 20);
```

Aynı anda birkaç kayıt ekleyebilirsin:

```sql
INSERT INTO ogrenci (ad, soyad, yas) VALUES ('Ayşe', 'Yılmaz', 22), ('Mehmet', 'Kara', 19);
```

---

**UPDATE**  
Var olan bir kaydı günceller.

```sql
UPDATE ogrenci SET yas = 21 WHERE ad = 'Ali' AND soyad = 'Vural';
```

Birden fazla alanı aynı anda güncelleyebilirsin:

```sql
UPDATE ogrenci SET yas = 22, soyad = 'Duran' WHERE id = 5;
```

---

**DELETE**  
Kayıt siler.

```sql
DELETE FROM ogrenci WHERE ad = 'Ali' AND soyad = 'Vural';
```

Belirli bir aralıktaki kayıtları silmek için:

```sql
DELETE FROM ogrenci WHERE yas < 18;
```

---

**TRUNCATE**  
Tablodaki tüm satırları hızlıca siler. Ama tabloyu korur.

```sql
TRUNCATE TABLE ogrenci;
```

---

**WHERE**  
Bir komutun sadece belirli satırlarda çalışmasını sağlar.

```sql
SELECT * FROM ogrenci WHERE yas > 18;
```

Birden fazla şart eklemek için **AND** ve **OR** kullanılır.

## Karşılaştırma ve Mantıksal Operatörler

SQL'de sorgulama yaparken aşağıdaki karşılaştırma ve mantıksal operatörleri kullanabilirsin:

| Operatör    | Açıklama                                                         |
|-------------|------------------------------------------------------------------|
| =           | Eşittir                                                          |
| <>          | Eşit değildir                                                    |
| >           | Büyüktür                                                         |
| <           | Küçüktür                                                         |
| >=          | Büyük veya eşittir                                               |
| <=          | Küçük veya eşittir                                               |
| **BETWEEN** | Arasındadır (belirli bir aralıkta olup olmadığını kontrol eder)   |
| **LIKE**    | İle başlar, ile biter, içerir anlamı taşır                       |
| **IN**      | İçindedir (verilen listedeki değerlerle eşleşir)                  |
| **NOT LIKE**| İle başlamaz, ile bitmez, içermez                                |
| **NOT IN**  | İçinde değildir (verilen listedeki değerlerde yoktur)             |

### Örnek Kullanımlar

```sql
SELECT * FROM ogrenci WHERE yas = 18;             -- Eşittir
SELECT * FROM ogrenci WHERE soyad <> 'Vural';     -- Eşit değildir
SELECT * FROM ogrenci WHERE yas > 18;             -- Büyüktür
SELECT * FROM ogrenci WHERE yas < 18;             -- Küçüktür
SELECT * FROM ogrenci WHERE yas >= 18;            -- Büyük veya eşittir
SELECT * FROM ogrenci WHERE yas <= 18;            -- Küçük veya eşittir
SELECT * FROM ogrenci WHERE yas BETWEEN 18 AND 22;-- 18 ile 22 arasında
SELECT * FROM ogrenci WHERE ad LIKE 'A%';         -- 'A' ile başlayan
SELECT * FROM ogrenci WHERE ad LIKE '%i';         -- 'i' ile biten
SELECT * FROM ogrenci WHERE ad LIKE '%li%';       -- İçinde 'li' geçen
SELECT * FROM ogrenci WHERE id IN (1,3,5);        -- id'si 1, 3 veya 5 olan
SELECT * FROM ogrenci WHERE ad NOT LIKE 'A%';     -- 'A' ile başlamayan
SELECT * FROM ogrenci WHERE id NOT IN (2,4,6);    -- id'si 2, 4, 6 olmayan
```

Bu operatörler ile sorgularını daha esnek ve güçlü şekilde yazabilirsin.





---

**AND ve OR**

AND ile birden fazla şartın hepsi sağlanmalı:

```sql
SELECT * FROM ogrenci WHERE yas > 18 AND soyad = 'Vural';
```

OR ile şartlardan biri yeterli:

```sql
SELECT * FROM ogrenci WHERE yas < 18 OR soyad = 'Vural';
```

Daha karmaşık şartlar için parantez kullanabilirsin:

```sql
SELECT * FROM ogrenci WHERE (yas > 18 OR soyad = 'Vural') AND ad = 'Ali';
```

---

**UPDATE ve DELETE Kullanırken WHERE**

WHERE kullanmazsan komut tablodaki tüm satırları etkiler. Sadece bir satırı değiştirmek veya silmek için WHERE ile özelleştir.

Bir satırı güncelle:

```sql
UPDATE ogrenci SET yas = 22 WHERE id = 5;
```

Bir satırı sil:

```sql
DELETE FROM ogrenci WHERE id = 3;
```

---

**DISTINCT**  
Tekrar eden değerleri bir kere gösterir.

```sql
SELECT DISTINCT yas FROM ogrenci;
```

Birden fazla sütunu beraber kullanabilirsin:

```sql
SELECT DISTINCT ad, soyad FROM ogrenci;
```

---

**ORDER BY**  
Sonuçları sıralar. Varsayılan olarak artan sıralama yapar.

Artan sırada:

```sql
SELECT * FROM ogrenci ORDER BY yas ASC;
```

Azalan sırada:

```sql
SELECT * FROM ogrenci ORDER BY yas DESC;
```

Birden fazla sütuna göre sıralama:

```sql
SELECT * FROM ogrenci ORDER BY soyad ASC, yas DESC;
```

---

**TOP (veya LIMIT)**  
Sadece belirli sayıda sonuç döndürür.

İlk 3 satırı almak için:

```sql
SELECT TOP 3 * FROM ogrenci;
```

Bazı SQL türlerinde aşağıdaki gibi de kullanılır:

```sql
SELECT * FROM ogrenci LIMIT 3;
```

---

#### 2. Veritabanı Yapısını Değiştirmek (DDL)

Veritabanı ve tabloların yapısını değiştiren komutlardır.

**CREATE**  
Tablo oluşturmak için kullanılır.

```sql
CREATE TABLE ogrenci (
    id INT PRIMARY KEY,
    ad VARCHAR(50) NOT NULL,
    soyad VARCHAR(50) NOT NULL,
    yas INT
);
```

Yeni bir veritabanı oluşturmak için:

```sql
CREATE DATABASE okul;
```

---

**ALTER**  
Var olan tabloyu değiştirmek için.

Yeni sütun eklemek:

```sql
ALTER TABLE ogrenci ADD dogum_tarihi DATE;
```

Bir sütunun tipini değiştirmek:

```sql
ALTER TABLE ogrenci ALTER COLUMN soyad VARCHAR(100);
```

Bir sütunu silmek:

```sql
ALTER TABLE ogrenci DROP COLUMN dogum_tarihi;
```

---

**DROP**  
Tabloyu veya veritabanını tamamen siler. Bu işlem geri alınamaz.

Tabloyu silmek için:

```sql
DROP TABLE ogrenci;
```

Veritabanını silmek için:

```sql
DROP DATABASE okul;
```

#### Sık Kullanılan Diğer SQL Komutları ve Fonksiyonları

**BETWEEN:** Değerin iki sayı arasında olup olmadığını kontrol eder.

```sql
SELECT * FROM ogrenci WHERE yas BETWEEN 18 AND 25;
```

**IN:** Değerin belirli bir listede olup olmadığını kontrol eder.

```sql
SELECT * FROM ogrenci WHERE ad IN ('Ali', 'Ayşe', 'Mehmet');
```

**LIKE:** Metinlerde benzerlik arar.

```sql
SELECT * FROM ogrenci WHERE ad LIKE 'A%';
```

**COUNT, SUM, AVG, MIN, MAX:**  
Sırasıyla toplam sayı, toplam değer, ortalama, en küçük ve en büyük değer hesaplar.

```sql
SELECT COUNT(*) FROM ogrenci;
SELECT AVG(yas) FROM ogrenci;
SELECT MAX(yas) FROM ogrenci;
```

**GROUP BY:**  
Kayıtları belirli bir alana göre gruplar.

```sql
SELECT yas, COUNT(*) FROM ogrenci GROUP BY yas;
```

**HAVING:**  
Gruplanmış kayıtlar üzerinde şart ekler.

```sql
SELECT yas, COUNT(*) FROM ogrenci GROUP BY yas HAVING COUNT(*) > 1;
```
