
# SQL Nedir?

SQL (Structured Query Language -Yapısal Sorgusal Dil), ilişkisel veritabanlarını yönetmek ve sorgulamak için kullanılan standart bir dildir.

## SQL'in Tarihçesi

SQL, 1970'lerde IBM'de Edgar F. Codd tarafından tanımlanan ilişkisel veritabanı modelini temel alır. SQL'in ilk hali, 1974 yılında Donald D. Chamberlin ve Raymond F. Boyce tarafından “SEQUEL” ismiyle geliştirilmiştir. Sonrasında ise isim “SQL” olarak değiştirilmiş ve bu dil hızla popülerleşmiştir. 1986 yılında ANSI ve 1987 yılında ISO tarafından standart olarak kabul edilmiştir.

## Neden SQL Kullanılır?

- **Veri Sorgulama ve Yönetme:** Veritabanındaki verileri eklemek, güncellemek, silmek ve sorgulamak için kullanılır.
- **Standartlaştırma:** Farklı veritabanı sistemlerinde (Oracle, Microsoft SQL Server, MySQL, PostgreSQL, vb.) ortak bir dil olarak kullanılabilir.
- **Veri Güvenliği ve Bütünlüğü:** Yetkilendirme ve veri bütünlüğü kuralları tanımlanabilir.
- **Verimlilik:** Büyük miktarda verinin hızlı ve etkili bir şekilde yönetilmesini sağlar.

Günümüzde hemen hemen tüm ilişkisel veritabanı sistemlerinde SQL kullanılmaktadır.


## Veri Tabanı

Veri tabanı, bilgilerin düzenli bir şekilde saklandığı ve gerektiğinde kolayca erişilebildiği bir yapıdır. Uygulamalar ve sistemler, verileri güvenli ve tutarlı bir şekilde depolamak için veri tabanlarını kullanır. Örneğin; kullanıcı kayıtları, satış verileri ya da ürün envanteri gibi bilgiler veri tabanlarında tutulur.

Veri tabanlarını yönetmek için genellikle bir veritabanı yönetim sistemi (DBMS) kullanılır. Bu sistemler sayesinde verilerin saklanması, düzenlenmesi, güvenliği ve bütünlüğü sağlanır. Yaygın olarak kullanılan ilişkisel veritabanı yönetim sistemleri arasında SQL Server, MySQL ve PostgreSQL bulunur.

Veri tabanlarında işlem yapmak ve veriyle ilgili sorgular oluşturmak için SQL (Structured Query Language) adlı özel bir dil kullanılır. SQL sayesinde veri ekleme, silme, güncelleme ve sorgulama işlemleri yapılabilir.

## Veri Tabanının Temel Bileşenleri

Bir veri tabanını oluşturan ana bileşenler şunlardır:

- **Tablo (Table):** Verilerin satırlar ve sütunlar halinde düzenli olarak saklandığı yapılardır. Her tablo, belirli bir veri türüne ait kayıtları içerir.
- **Satır (Row/Kayıt):** Tabloya eklenen her bir veri parçasını temsil eder; her satır bir kayıttır.
- **Sütun (Column):** Tablo içerisindeki verilerin türünü ve özelliklerini belirler. Her sütun, bir veri alanını gösterir.
- **Birincil Anahtar (Primary Key):** Her kaydı benzersiz şekilde tanımlayan sütundur. Aynı değere sahip iki satır olamaz.
- **Yabancı Anahtar (Foreign Key):** Bir tablodaki satırı başka bir tablodaki satırla ilişkilendiren sütundur. Tablo ilişkilerinin kurulmasını sağlar.
- **İndeks (Index):** Verilere hızlı erişim sağlamak amacıyla oluşturulan özel veri yapılarıdır.

Bu temel bileşenler, ilişkisel veri tabanlarının yapısını oluşturur ve verilerin düzenli, kolayca yönetilebilir, güvenli bir şekilde depolanmasına yardımcı olur.


### **Veritabanı Yönetim Sistemi (DBMS - Database Management System):**

Veritabanı yönetim sistemi, verilerin bilgisayar ortamında düzenli, güvenli ve tutarlı bir şekilde saklanmasını ve yönetilmesini sağlayan yazılımlardır. DBMS ile verileri ekleyebilir, silebilir, güncelleyebilir ve sorgulayabilirsiniz. Ayrıca, veritabanı üzerindeki kullanıcı yetkilerini yönetme, veri bütünlüğünü sağlama ve veri güvenliğini koruma gibi görevleri yerine getirir. Popüler DBMS örnekleri arasında Microsoft SQL Server, MySQL, Oracle ve PostgreSQL sayılabilir.


### **Veritabanı Sunucusu:**
Veritabanı sunucusu ise, DBMS yazılımının çalıştığı ve veritabanlarının fiziksel olarak barındırıldığı bilgisayardır. Sunucu üzerinde, bir veya birden fazla veritabanı bulunabilir ve ağ üzerinden gelen istekler bu sunucuya iletilerek veriler sorgulanır ya da güncellenir. Kısacası veritabanı sunucusu, verilerin bulunduğu ve uygulamaların bu verilere erişmesini sağlayan bilgisayar veya sunucu sistemidir.

## İlişkisel Veritabanı Nedir?

İlişkisel veritabanı, verilerin tablolar (satırlar ve sütunlar) şeklinde düzenli olarak saklandığı bir veritabanı modelidir. Bu modelde:

- **Veriler, ilişkili tablolarda saklanır.** Her tablo, belirli bir veri türüne ait kayıtların bulunduğu yapıdır.
- **Tablolar arasındaki bağlantılar anahtarlar** (primary key ve foreign key) aracılığıyla kurulur.
- **Verilerin tutarlılığı ve bütünlüğü korunur**, redundant veri tekrarı önlenir.
- Farklı tablolardaki veriler kolayca birleştirilebilir ve sorgulanabilir.

Örneğin; bir "Müşteriler" tablosu ile bir "Siparişler" tablosu, müşteri kimliği üzerinden birbirine bağlanabilir. Böylece, bir müşterinin yaptığı tüm siparişler kolayca sorgulanabilir.

İlişkisel veritabanı sayesinde büyük ve karmaşık veri kümeleri yönetilebilir, esnek sorgularla veri analizleri yapılabilir.
