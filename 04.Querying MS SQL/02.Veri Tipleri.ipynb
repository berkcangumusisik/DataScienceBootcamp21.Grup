## MSSQL Tam Sayı (Integer) Tipleri

Microsoft SQL Server'da tam sayı (integer) veri tipleri, tam sayıları saklamak için kullanılır ve farklı boyutlarda çeşitli türler sunar. İşte başlıca MSSQL tam sayı türleri:

### 1. `BIT`
- **Depolama Alanı:** 1 bit
- **Alabileceği Değerler:** 0 (yanlış/false) veya 1 (doğru/true)
- **Kullanım Alanı:** Genellikle doğru/yanlış (mantıksal) değerlerin saklanması için kullanılır.

### 2. `TINYINT`
- **Depolama Alanı:** 1 bayt (8 bit)
- **Alabileceği Değer Aralığı:** 0 ile 255 (işaretsiz)
- **Kullanım Alanı:** Küçük pozitif tamsayı verileri için uygundur.

### 3. `SMALLINT`
- **Depolama Alanı:** 2 bayt (16 bit)
- **Alabileceği Değer Aralığı:** -32,768 ile 32,767
- **Kullanım Alanı:** Küçük pozitif ve negatif tamsayılar için kullanılır.

### 4. `INT`
- **Depolama Alanı:** 4 bayt (32 bit)
- **Alabileceği Değer Aralığı:** -2,147,483,648 ile 2,147,483,647
- **Kullanım Alanı:** Standart tamsayı verileri için en sık kullanılan veri tipidir.

### 5. `BIGINT`
- **Depolama Alanı:** 8 bayt (64 bit)
- **Alabileceği Değer Aralığı:** -9,223,372,036,854,775,808 ile 9,223,372,036,854,775,807
- **Kullanım Alanı:** Çok büyük tamsayı değerleri için kullanılır.

### Özet Tablo

| Veri Tipi  | Bayt       | Minimum Değer                | Maksimum Değer                |
|------------|------------|------------------------------|-------------------------------|
| BIT        | ~0.125     | 0                            | 1                             |
| TINYINT    | 1          | 0                            | 255                           |
| SMALLINT   | 2          | -32,768                      | 32,767                        |
| INT        | 4          | -2,147,483,648               | 2,147,483,647                 |
| BIGINT     | 8          | -9,223,372,036,854,775,808   | 9,223,372,036,854,775,807     |

**Not:** MSSQL'de tam sayı tiplerinin seçimi, saklanacak verinin minimum ve maksimum değer aralığına ve performans ihtiyacına göre yapılmalıdır. `BIT` türü, sadece iki değer (0 veya 1) saklanacaksa tercih edilir ve mantıksal işlemler için oldukça uygundur.

## MSSQL String (Karakter) Veri Tipleri

Microsoft SQL Server'da karakter (string) verilerini saklamak için çeşitli veri tipleri kullanılır. Bu tipler, metnin uzunluğuna ve içerdiği karakterlerin yapısına göre seçilir. Temelde iki ana kategori vardır: sabit uzunluklu ve değişken uzunluklu karakter tipleri. Ayrıca Unicode karakterleri destekleyen tipler de bulunur.

### 1. `CHAR(n)`
- **Tanım:** Sabit uzunluklu karakter dizisi.
- **Depolama Alanı:** Her zaman `n` karakterlik alan kaplar (1 ≤ n ≤ 8.000).
- **Kullanım Alanı:** Uzunluğu her zaman aynı olacak metinler için uygundur (örneğin, kodlar).

### 2. `VARCHAR(n)`
- **Tanım:** Değişken uzunluklu karakter dizisi.
- **Depolama Alanı:** Girilen karakter sayısı kadar alan kaplar (1 ≤ n ≤ 8.000).
- **Kullanım Alanı:** Uzunluğu değişebilen metinler için kullanılır.

### 3. `VARCHAR(MAX)`
- **Tanım:** 8.000 karakterden uzun metinleri depolamak için kullanılır.
- **Depolama Alanı:** 2^31-1 (yaklaşık 2 GB) karaktere kadar veri saklayabilir.
- **Kullanım Alanı:** Çok uzun metinler, açıklamalar veya doküman içerikleri için uygundur.

### 4. `NCHAR(n)`
- **Tanım:** Sabit uzunluklu Unicode karakter dizisi.
- **Depolama Alanı:** Her karakter için 2 bayt, toplamda 2n bayt yer kaplar (1 ≤ n ≤ 4.000).
- **Kullanım Alanı:** Sabit uzunlukta, çoklu dil (Unicode) desteği gereken alanlarda kullanılır.

### 5. `NVARCHAR(n)`
- **Tanım:** Değişken uzunluklu Unicode karakter dizisi.
- **Depolama Alanı:** Her karakter için 2 bayt yer kaplar (1 ≤ n ≤ 4.000).
- **Kullanım Alanı:** Farklı dillerde karakterlerin saklanması gerektiğinde ve metin uzunluğu değişkense tercih edilir.

### 6. `NVARCHAR(MAX)`
- **Tanım:** Çok uzun Unicode metinler için kullanılır.
- **Depolama Alanı:** 2^31-1 karaktere kadar veri depolanabilir.
- **Kullanım Alanı:** Çok büyük açıklamalar, içerik ve dokümanlar için uygundur.

### Özet Tablo

| Veri Tipi     | Maksimum Uzunluk     | Unicode Desteği | Sabit/Değişken  | Kullanım Amacı                                     |
| ------------- | -------------------- | --------------- | --------------- | -------------------------------------------------- |
| CHAR(n)       | 8.000 karakter       | Hayır           | Sabit           | Sabit uzunluklu kısa metinler                      |
| VARCHAR(n)    | 8.000 karakter       | Hayır           | Değişken        | Kısa/orta, değişken uzunluklu metinler             |
| VARCHAR(MAX)  | 2^31-1 karakter      | Hayır           | Değişken        | Çok uzun metinler                                  |
| NCHAR(n)      | 4.000 karakter       | Evet            | Sabit           | Sabit uzunluklu, çoklu dil desteği gerektiren alanlar|
| NVARCHAR(n)   | 4.000 karakter       | Evet            | Değişken        | Değişken uzunluklu, çoklu dil destekli metinler    |
| NVARCHAR(MAX) | 2^31-1 karakter      | Evet            | Değişken        | Çok uzun, çoklu dil destekli metinler              |

**Not:** Unicode desteği gerektiren bilgiler (örneğin Türkçe karakterler veya farklı alfabeler) için `NCHAR` ve `NVARCHAR` tipleri; yalnızca İngilizce karakterlerle sınırlı veriler için ise `CHAR` ve `VARCHAR` tipleri tercih edilmelidir. Uzunluğu değişebilen verilerde gereksiz veri kaybı ve fazladan alan kullanımını önlemek için değişken uzunluklu tipler daha uygundur.

## MSSQL Ondalık (Decimal) Sayı Veri Tipleri

Microsoft SQL Server'da ondalıklı (küsuratlı) sayıları saklamak için çeşitli veri tipleri bulunur. Bu tipler, hassasiyet ve ölçek gereksinimlerine göre seçilir. Genellikle finansal veriler ve kesin matematiksel hesaplamalar için kullanılır.

### 1. `DECIMAL(p, s)` / `NUMERIC(p, s)`
- **Tanım:** Sabit hassasiyet ve ölçeğe sahip sayısal veri tipleridir. `DECIMAL` ve `NUMERIC` aynı işlevi görür.
- **Parametreler:**
  - `p` (Precision): Toplam rakam sayısı (1-38 arası).
  - `s` (Scale): Ondalık noktadan sonraki rakam sayısı (0 ≤ s ≤ p).
- **Kullanım Alanı:** Finansal uygulamalar, para birimleri, kesin aritmetik işlemler.
- **Örnek:** `DECIMAL(10, 2)` ifadesiyle toplam 10 rakam, bunun 2'si virgülden sonra tutulabilir: `12345678.90`

### 2. `FLOAT(n)`
- **Tanım:** Yaklaşık değere sahip kayan noktalı sayı tipidir. `n` bit cinsinden kesinliği ifade eder.
- **Depolama:**
  - `n` 1-24 arası ise 4 bayt (`FLOAT(24)` ≈ `REAL`)
  - `n` 25-53 arası ise 8 bayt
- **Kullanım Alanı:** Bilimsel ve mühendislik uygulamalarında yaklaşık sonuçlar yeterliyse kullanılır.
- **Örnek:** `FLOAT(53)`

### 3. `REAL`
- **Tanım:** 4 baytlık kayan noktalı (yaklaşık) sayı tipidir. `FLOAT(24)` ile eşdeğerdir.
- **Kullanım Alanı:** Hassasiyetin çok önemli olmadığı ve az bellek kullanılması istenen durumlarda tercih edilir.

### Özet Tablo

| Veri Tipi       | Tanım                     | Maksimum Hassasiyet | Kullanım Amacı                                   |
|-----------------|-------------------------- |---------------------|--------------------------------------------------|
| DECIMAL(p, s)   | Sabit ondalık             | 38 basamak          | Finans, para, hassas matematik                   |
| NUMERIC(p, s)   | Sabit ondalık             | 38 basamak          | Finans, para, hassas matematik                   |
| FLOAT(n)        | Kayan nokta (yaklaşık)    | 53 basamak          | Bilimsel hesaplar, yaklaşık sonuçlar             |
| REAL            | Kayan nokta (yaklaşık)    | 24 basamak          | Düşük bellek, yaklaşık sonuçlar                  |

**Not:**
- Kesin sonuçlar ve para işlemleri için `DECIMAL` veya `NUMERIC` tipleri tercih edilmelidir.
- `FLOAT` ve `REAL` tipleri yaklaşık değer sağladığı için yüksek hassasiyet gereken işlemlerde önerilmez.

## MSSQL Tarih ve Saat (Date & Time) Veri Tipleri

Microsoft SQL Server'da tarih ve saat verilerini saklamak için çeşitli veri tipleri kullanılır. Bu tipler, uygulamanın ihtiyaç duyduğu hassasiyet ve kullanım amacına göre seçilir.

### 1. `DATE`
Yalnızca yılı, ayı ve günü saklar (YYYY-MM-DD formatında). Zaman bilgisi içermez.
Örnek: `2023-05-21`

### 2. `TIME(p)`
Sadece saat, dakika, saniye ve isteğe bağlı olarak saniyenin kesirli (hassasiyetli) kısmını saklar. `p` ile hassasiyet belirlenir (0-7 arası).
Örnek: `14:30:45.1234567`

### 3. `DATETIME`
Hem tarih hem saat (gün, ay, yıl, saat, dakika, saniye ve 1/300 saniye hassasiyet) bilgisini tutar. Sıkça kullanılır; 1753-9999 yılları arasını destekler.
Örnek: `2023-05-21 14:30:45`

### 4. `SMALLDATETIME`
Daha az alan kaplar; dakika hassasiyetindedir. 1900-2079 yılları arasını destekler.
Örnek: `2023-05-21 14:30`

### 5. `DATETIME2(p)`
`DATETIME`'a benzer ama daha hassas ve daha geniş tarih aralığına sahip (0001-9999 yılları arası, 0-7 arası saniye hassasiyeti seçilebilir).
Örnek: `2023-05-21 14:30:45.1234567`

### 6. `DATETIMEOFFSET(p)`
`DATETIME2` ile aynıdır, ek olarak zaman dilimi farkını (+03:00, -05:00 gibi) saklar. Zamanın farklı bölgelerde hesaplanmasında kullanılır.
Örnek: `2023-05-21 14:30:45.1234567 +03:00`

#### Özelliklerin Karşılaştırıldığı Tablo

| Veri Tipi         | Saklanan Bilgi          | Tarih Aralığı         | Zaman Hassasiyeti    | Zaman Dilimi     | Alan (bayt)   | Kullanım Amacı                                  |
|-------------------|------------------------|-----------------------|----------------------|------------------|---------------|------------------------------------------------|
| DATE              | Yıl, ay, gün           | 0001-9999             | -                    | Hayır            | 3             | Sadece tarih saklama                            |
| TIME(p)           | Saat, dakika, saniye   | -                     | 0-7 basamak          | Hayır            | 3-5           | Sadece saat saklama                             |
| DATETIME          | Tarih, saat            | 1753-9999             | 1/300 sn             | Hayır            | 8             | Genel tarih ve saat, geri uyumluluk             |
| SMALLDATETIME     | Tarih, saat            | 1900-2079             | Dakika               | Hayır            | 4             | Kısıtlı aralık/küçük yer kaplama                |
| DATETIME2(p)      | Tarih, saat            | 0001-9999             | 0-7 basamak          | Hayır            | 6-8           | Yüksek doğruluk, geniş aralık                   |
| DATETIMEOFFSET(p) | Tarih, saat, zaman farkı| 0001-9999            | 0-7 basamak          | Evet             | 8-10          | Farklı zaman dilimleriyle çalışmak              |

Not:
- `DATE` ve `TIME` veri tipleri, yalnızca tarih veya yalnızca saat gerektiren durumlarda alan tasarrufu sağlar.
- Uygulamanız zaman dilimi ile ilgiliyse `DATETIMEOFFSET` tercih edilmelidir.
