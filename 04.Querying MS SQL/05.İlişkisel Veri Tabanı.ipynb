## İlişkisel Veri Tabanı (Relational Database) Nedir?

İlişkisel veri tabanları, verilerin birbiriyle ilişkili tablolar halinde depolandığı, en yaygın kullanılan veri tabanı sistemleridir. Her tablo, belirli bir konuya veya varlık türüne ait verileri düzenli olarak saklar; örneğin, "Müşteriler" veya "Siparişler" gibi. Tablolar satırlardan (her satır bir veri kaydını ifade eder) ve sütunlardan (her sütun bir veri alanını gösterir) oluşur.

İlişkisel veri tabanlarında, tablolar arasındaki ilişkiler anahtarlar (key) aracılığıyla kurulur. Her tabloda genellikle bir “birincil anahtar” (primary key) sütunu bulunur ve diğer tablolardaki “yabancı anahtarlar” (foreign key) ile bağlantı kurulur. Böylelikle veri tekrarından kaçınılır ve veri bütünlüğü sağlanır.

### İlişkisel Veri Tabanlarının Temel Özellikleri

- Veriler, birden çok tabloya ayrılarak saklanır.
- Tablolar arasında anahtarlar sayesinde ilişki kurulabilir.
- Tekrarlı veri depolamanın önüne geçilir.
- Veri bütünlüğü ve tutarlılığı sağlanır.
- SQL gibi sorgu dilleriyle kolayca veri çekme ve analiz işlemleri yapılabilir.

## Join (Birleştirme) Nedir?

Join işlemi, ilişkisel veri tabanlarındaki farklı tablolardan, ortak sütunlar aracılığıyla anlamlı veriler elde etmek için kullanılır. Özellikle birden fazla tablo arasında bağlantılı bilgiye ihtiyaç duyulduğunda SQL’in JOIN komutları ile tablolar birleştirilebilir.

### Başlıca Join Türleri

- **INNER JOIN:** Sadece her iki tabloda da eşleşen kayıtları döndürür.
- **LEFT JOIN (LEFT OUTER JOIN):** Sol tablodaki tüm kayıtları ve buna karşılık gelen sağ tablodaki eşleşen kayıtları döndürür. Eşleşmeyen sağ taraf kayıtları NULL olur.
- **RIGHT JOIN (RIGHT OUTER JOIN):** Sağ tablodaki tüm kayıtları ve sol tablodaki eşleşenleri döndürür. Eşleşmeyen sol taraf kayıtları NULL olur.
- **FULL JOIN (FULL OUTER JOIN):** Her iki tablodaki tüm kayıtları döndürür. Eşleşmeyen taraf için NULL değerler gösterilir.

## Kullanıma Yönelik Örnek

Diyelim ki elimizde iki tablo mevcut: `Musteriler` ve `Siparisler`.

- `Musteriler` tablosu: MusteriID, Ad, Soyad
- `Siparisler` tablosu: SiparisID, MusteriID, Tarih

Bir müşterinin yaptığı siparişleri birlikte listelemek için INNER JOIN kullanabiliriz:

```sql
SELECT Musteriler.Ad, Musteriler.Soyad, Siparisler.Tarih
FROM Musteriler
INNER JOIN Siparisler ON Musteriler.MusteriID = Siparisler.MusteriID
```

Bu sorgu sayesinde, hem müşteri bilgilerini hem de bu müşteriye ait sipariş tarihlerini görebilirsiniz. Join işlemleri sayesinde tablolar arasındaki tüm ilişkili bilgileri esnek ve etkili bir şekilde sorgulayarak büyük veri kümelerinde anlamlı analizler yapabilirsiniz.

İlk kez kullanan biri olarak, ilişkisel veri tabanlarında tabloları ve join işlemlerini kavradığınızda, verilerinizi çok daha iyi yönetebilir ve ilişkili bilgileri tek bir sorguda kolayca birleştirebilirsiniz.

## Alt Sorgu (Subquery) Nedir?

Alt sorgu (subquery), bir SQL sorgusu içinde başka bir SQL sorgusu olarak kullanılan yapıdır. Alt sorgular genellikle SELECT, INSERT, UPDATE veya DELETE ifadelerinde; WHERE, FROM veya SELECT gibi farklı yerlerde kullanılabilir. Alt sorgu sonuçları, ana sorgunun filtresinde veya veri toplamada kullanılabilir.

### Alt Sorguya Örnek Kullanım

Diyelim ki bir `Siparisler` tablomuz ve bir de `Musteriler` tablomuz var. Ve sadece “Ankara” ilinde bulunan müşterilerin verdiği siparişleri listelemek istiyoruz.

- `Musteriler` tablosu: MusteriID, Ad, Sehir
- `Siparisler` tablosu: SiparisID, MusteriID, Tarih

Şu sorgu ile alt sorgu kullanılabilir:

```sql
SELECT *
FROM Siparisler
WHERE MusteriID IN (
    SELECT MusteriID
    FROM Musteriler
    WHERE Sehir = 'Ankara'
)
```

Burada, içteki alt sorgu (`SELECT MusteriID FROM Musteriler WHERE Sehir = 'Ankara'`) Ankara’daki müşterilerin ID’lerini bulur; ana sorgu ise bu ID’lerin geçmiş olduğu siparişleri getirir.

Alt sorgular, karmaşık veri işlemlerini basit ve okunabilir hale getirirken, mantıksal olarak birden fazla adımda karar vermenizi sağlar. Büyük veri tabanlarında etkili analizler ve dinamik filtrelemeler için sıkça kullanılır.
