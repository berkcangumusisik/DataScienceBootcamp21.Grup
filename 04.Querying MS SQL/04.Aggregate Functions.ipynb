## MSSQL'de Aggregate Functions (Toplama Fonksiyonları) ve GROUP BY Kullanımı

### Aggregate Functions Nedir?
Aggregate Functions (Toplama/Sayısal Fonksiyonlar), bir grup veri üzerinde toplu hesaplamalar yapmanızı sağlayan fonksiyonlardır. Genellikle birden fazla satırı tek bir değer ile özetlemek için kullanılır.

En sık kullanılan aggregate function'lar şunlardır:

- **COUNT()**: Satırları sayar.
- **SUM()**: Bir sütundaki değerlerin toplamını alır.
- **AVG()**: Bir sütundaki değerlerin ortalamasını hesaplar.
- **MIN()**: Bir sütundaki en küçük değeri bulur.
- **MAX()**: Bir sütundaki en büyük değeri bulur.

#### Örnek Kullanımlar
```sql
SELECT COUNT(*) FROM Customers;
SELECT SUM(Price) FROM Orders;
SELECT AVG(Age) FROM Employees;
SELECT MIN(Salary) FROM Employees;
SELECT MAX(OrderDate) FROM Orders;
```

### GROUP BY Nedir?
`GROUP BY` ifadesi, tablodaki verileri belirli sütunlara göre gruplandırmak ve bu gruplara aggregate function uygulamak için kullanılır.

#### Temel Söz Dizimi
```sql
SELECT kolon_adi, AGGREGATE_FUNCTION(kolon_adi)
FROM tablo_adi
GROUP BY kolon_adi;
```

#### Örnek:
Bir şirketin farklı departmanlarındaki toplam maaşı listelemek için:
```sql
SELECT Department, SUM(Salary) AS TotalSalary
FROM Employees
GROUP BY Department;
```
Bu sorgu, her bir departman için toplam maaşı hesaplayacaktır.

### Notlar
- `GROUP BY` genellikle aggregate function'larla beraber kullanılır.
- `SELECT` ifadesinde, `GROUP BY`da olmayan her sütun bir aggregate function içinde olmalıdır.
- Sonuçları filtrelemek için `HAVING` ile aggregate fonksiyonlara istediğiniz koşulları ekleyebilirsiniz.

#### HAVING Kullanımı Örneği:
```sql
SELECT Department, COUNT(*) AS EmployeeCount
FROM Employees
GROUP BY Department
HAVING COUNT(*) > 5;
```
Bu sorgu, yalnızca 5'ten fazla çalışanı olan departmanları getirir.

---

Aggregate Functions ve GROUP BY, büyük veri kümelerinde özet bilgi çıkarmak için güçlü araçlardır.
