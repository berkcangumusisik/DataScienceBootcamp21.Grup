Aykırı Değerler (Outliers) Nedir?
---------------------------------

**Aykırı değerler**, bir veri kümesindeki çoğu değerden çok farklı olan, uçta kalan veri noktalarıdır. Yeni başlayanlar için şöyle açıklayabiliriz: Eğer elinizde insanların boyları verisi varsa, 200 cm'nin çok üstünde veya 100 cm'nin çok altında bir değer görürseniz, bu "aykırı değer" olabilir.

Neden Önemlidir?
----------------

- Aykırı değerler yanlış ölçümlerden veya sıra dışı durumlardan kaynaklanabilir.
- Makine öğrenmesi modelleri, bu uç değerler nedeniyle yanlış öğrenmeler yapabilir ya da kötü tahminlerde bulunabilir.
- Aykırı değerleri doğru tespit edip incelemek, daha doğru sonuçlar elde etmemizi sağlar.

Aykırı Değerler Nasıl Bulunur?
------------------------------

En sık kullanılan yöntemlerden biri, veriyi **görselleştirmek** ve istatistiksel teknikler kullanmaktır. Özellikle kutu grafiği (boxplot), aykırı değerleri tespit etmek için çok işe yarar.


Aykırı Değerlerle Ne Yapmalı?
-----------------------------

- **Kontrol et:** Bu değerler bir hata mı, yoksa gerçek bir gözlem mi?
- **Düzelt veya çıkar:** Hatalıysa düzeltebilir ya da veri setinden çıkarabilirsin.
- **Ayrı analiz et:** Bazen aykırı değerler analiz için önemli olabilir (örneğin dolandırıcılık tespiti).

Kısacası, aykırı değerler makine öğrenmesinde ve veri analizinde önemli bir konudur. Onları tanımak ve doğru şekilde yönetmek her zaman gereklidir!

### Aykırı Değerleri Belirleme Yöntemleri

1. **Sektör Bilgisi (Domain Knowledge):**  
   Sektör veya alan uzmanlarının bilgisine başvurarak gözle batan uç değerlerin incelenmesi uygulanır. Örneğin, beklenmedik derecede yüksek bir veri gözlendiğinde sektör bilgisinden yararlanılarak bunun olağan dışı olup olmadığına karar verilir.

2. **Standart Sapma Yaklaşımı:**  
   Eğer veri normal dağılıyorsa, ortalamadan belirli bir standart sapma uzaklığında kalan değerler aykırı olarak kabul edilebilir. Genellikle 2 veya 3 standart sapmadan uzak olan gözlemler aykırı değer olarak değerlendirilir.

3. **Z-Skoru Yaklaşımı:**  
   Her bir veri noktasının ortalamadan kaç standart sapma uzaklıkta olduğunu hesaplamak için Z-skoru kullanılır. Z-skorunun mutlak değeri tipik olarak 3’ten büyükse, ilgili veri noktası aykırı olarak kabul edilir.

4. **Boxplot (Interquartile Range - IQR) Yöntemi:**  
   Kutu grafiği yardımıyla alt ve üst çeyrekler aralığına (IQR) bakılır. Birinci çeyreğin 1.5*IQR altına veya üçüncü çeyreğin 1.5*IQR üstüne düşen değerler aykırı değer olarak belirlenir.

Ek olarak, **LOF (Local Outlier Factor)** gibi gelişmiş yöntemler de özellikle çok boyutlu veri setlerinde aykırı değerlerin tespitinde kullanılır. LOF, veri noktalarının yoğunluklarını kıyaslayarak normalden sapmış olanları ortaya çıkarır ve makine öğrenmesi uygulamalarında sıklıkla tercih edilir.

Boxplot (Kutu Grafiği) Nedir?
-----------------------------

Boxplot, yani kutu grafiği; bir veri setinin dağılımını, merkezi eğilimini, yayılımını ve olası aykırı (uç) değerlerini hızlı ve etkin şekilde görselleştiren çok işlevli bir istatistiksel grafiktir. Özellikle veri analizi ve makine öğrenmesi süreçlerinde verinin genel özetini sunmak amacıyla sık kullanılır.

![Boxplot Açıklama Görseli](https://miro.medium.com/v2/resize:fit:1298/1*uj6Ej9LXTL-fOcbuDqlkjg.png)

### Boxplot’ın Temel Bileşenleri

1. **Minimum Değer (Alt Çizgi/Whisker)**  
   Aykırı değerler hariç, veri kümesinin en küçük değerini gösterir.

2. **Birinci Çeyrek (Q1)**  
   Verinin %25’i bu değerin altındadır. Kutu grafiğinde kutunun alt kenarıdır.

3. **Medyan (Q2)**  
   Veri setinin ortanca (medyan) değeridir. Kutunun ortasındaki çizgiyle gösterilir ve veri dağılımının ortasını temsil eder.

4. **Üçüncü Çeyrek (Q3)**  
   Verinin %75’i bu değerin altındadır. Kutu grafiğinde kutunun üst kenarıdır.

5. **Maksimum Değer (Üst Çizgi/Whisker)**  
   Aykırı değerler dışında kalan en yüksek veri noktasıdır.

6. **Aykırı Değerler (Outliers)**  
   Kutu ve çizgiler (whisker) dışında kalan ve genellikle nokta olarak gösterilen, alışılmadık derecede düşük veya yüksek değerlerdir.

### Boxplot ile Hangi Bilgileri Edinebiliriz?

- Verinin merkezi eğilimi (medyan)
- Dağılımın yaygınlığı (çeyrekler arası aralık – IQR: Q3-Q1)
- Olası aykırı değerler
- Verinin simetrik ya da çarpık olup olmadığı
- Minimum ve maksimum değerlerin aralığı

### Grafik Üzerindeki Açıklamalar ve Yorumlar

Yukarıda eklenen görselde, boxplot’ın tüm temel bölümleri işaretlenmiş durumdadır:

- Kutu (box): Q1 ve Q3 arası aralığı gösterir. Kutunun içinde yatay bir çizgiyle medyan (Q2) belirtilir.
- Whisker (çizgiler): Kutunun üstünden ve altından dışarı taşan çizgiler, aykırı (uç) değerler haricindeki minimum ve maksimum değerleri belirtir.
- Noktalar veya daireler: Kutu ve whisker’ların dışında kalan, 1.5 IQR’ın ötesinde yer alan veri noktalarıdır ve aykırı değerleri temsil eder.
- Grafik üzerinde ayrıca IQR (Interquartile Range, yani Q3-Q1) ve 1.5 × IQR hesaplamalarının aykırı değerleri nasıl tanımladığı açıklanır.

````python
# Aykırı değerler, bir değişkenin gözlemlerinin olağan dağılımının dışında kalan uç ya da sıra dışı değerlerdir. 
# Boxplot (kutu grafiği) çizildikten sonra, aykırı değerlerin sınırlarını belirlemek için çeyreklikler ve IQR kullanılır:
# Q1 (birinci çeyrek), verilerin %25'inin altında kalan değerdir.
# Q3 (üçüncü çeyrek), verilerin %75'inin altında kalan değerdir.
# IQR (Interquartile Range), Q3 ve Q1 arasındaki farktır ve veri setindeki yayılımı gösterir. 
q1 = df["Age"].quantile(0.25)      # Birinci çeyreklik (25. yüzdelik)
q3 = df["Age"].quantile(0.75)      # Üçüncü çeyreklik (75. yüzdelik)
iqr = q3 - q1                      # IQR, yani iki çeyreklik arasındaki fark
up = q3 + 1.5 * iqr               # Üst sınır
low = q1 - 1.5 * iqr              # Alt sınır
df[(df["Age"] < low) | (df["Age"] > up)]
# Bu kod, df veri çerçevesindeki "Age" sütununda, 1.5*IQR'ın altında veya üstünde olan değerleri filtreler. Yani bu değerler aykırı değerlerdir.
````

````python
def outlier_thresholds(dataframe, col_name):
    """
    Verilen bir dataframe ve değişken ismi için aykırı değer sınırlarını (alt ve üst limit) hesaplar.
    Alt ve üst limitleri çeyreklikler ve IQR (interquartile range) kullanarak belirler.
    
    Parameters:
        dataframe (pd.DataFrame): İşlem yapılacak veri çerçevesi.
        col_name (str): Sınırları hesaplanacak değişkenin ismi.
        
    Returns:
        tuple: (alt limit, üst limit)
    """
    quartile1 = dataframe[col_name].quantile(0.25)
    quartile3 = dataframe[col_name].quantile(0.75)
    interquantile_range = quartile3 - quartile1
    low_limit = quartile1 - 1.5 * interquantile_range
    up_limit = quartile3 + 1.5 * interquantile_range
    return low_limit, up_limit
outlier_thresholds(df, "Age")
````

````python
# Bir değişkende aykırı değer (outlier) olup olmadığını kontrol eden fonksiyonun sıfırdan açıklaması ve yazılması:

def check_outlier(dataframe, col_name):
    """
    Verilen veri çerçevesinde (dataframe) ve belirtilen değişkende (col_name) aykırı değer (outlier) olup olmadığını kontrol eder.
    Aykırı değerler, çeyreklikler arası mesafe (IQR) yöntemiyle saptanır.
    Eğer en az bir aykırı değer varsa True, yoksa False döner.

    Parametreler:
        dataframe (pd.DataFrame): Kontrol edilecek veri çerçevesi.
        col_name (str): Kontrol edilecek sütunun adı.

    Dönüş:
        bool: Aykırı değer varsa True, yoksa False.
    """
    # Alt ve üst sınırı hesapla
    low_limit, up_limit = outlier_thresholds(dataframe, col_name)
    
    # Aykırı değerleri bul: sütundaki değerlerden sınırların dışında kalanları filtrele
    outliers = dataframe[(dataframe[col_name] < low_limit) | (dataframe[col_name] > up_limit)]
    
    # En az 1 aykırı değer varsa True döndür, yoksa False döndür
    return outliers.shape[0] > 0

# Fonksiyonu örnek olarak "Fare" sütununda kullan:
check_outlier(df, "Fare")
````

````python
def grab_col_names(dataframe, cat_th=10, car_th=20):
    """
    Bir veri setindeki sütunları türlerine göre kategorize etmek, veriyi anlamak ve modelleme sürecinde doğru şekilde kullanmak için önemlidir.
    Bu fonksiyon, verilen bir pandas DataFrame'deki değişkenleri (sütunları) üç ana grupta toplar:
        1. Kategorik değişkenler (cat_cols)
        2. Numerik değişkenler (num_cols)
        3. Kategorik görünümlü ama çok fazla benzersiz değer içeren (kardinal) değişkenler (cat_but_car)
    Özellikle, veri analizine yeni başlayan birisi için, bu fonksiyonu şu şekilde adım adım anlatabiliriz:

    1. Kategorik değişkenler genellikle yazısal (string/obje tipinde) olur. Ancak bazen, içerisinde az sayıda benzersiz değer olan sayısal tipteki sütunlar da aslında kategoriktir. Örneğin, "1", "2", "3" gibi kodlarla ifade edilen bir sütun öğrencinin sınıfı olabilir.
    2. Numerik değişkenler sayısal değerler içerir ("int" veya "float" tipinde).
    3. Kategorik görünümlü kardinal değişkenler ise, yazısal olup çok fazla benzersiz değer içerir (örneğin müşteri ID'si). Bunları ayrı ele almak gerekir, çünkü makine öğrenmesi uygulamalarında bu tip sütunların doğrudan kullanılması uygun değildir.

    Fonksiyonun parametreleri:
        dataframe (pd.DataFrame): Analiz edilecek veri seti
        cat_th (int): Sayısal olup az sayıda benzersiz kategoriye sahip olan sütunların kategorik olarak kabul edilmesi için eşik değeri. Default: 10
        car_th (int): Kategorik olup çok fazla benzersiz değere sahip sütunların (kardinallerin) belirlenmesi için eşik değeri. Default: 20

    Fonksiyonun çıktıları:
        cat_cols (list): Kategorik sütun isimleri
        num_cols (list): Numerik sütun isimleri
        cat_but_car (list): Kategorik görünüp çok fazla benzersiz değer içeren sütun isimleri

    Fonksiyonun işleyişi:
        - Kategorik sütunları bulur ("object" tipinde olanlar ve az sayıda eşsiz değeri olan sayısal sütunlar).
        - Kategorik görünümlü kardinal sütunları bulur.
        - Kategorik türlere eklenenlerden kardinal olanları çıkarır.
        - Numerik sütunları bulur, ama az kategorili olan ("num_but_cat") sütunları hariç tutar.
        - Son olarak özet bilgiler yazdırır ve üç tür sütun ismini döndürür.

    Kullanım örneği:
        import seaborn as sns
        df = sns.load_dataset("iris")
        cat_cols, num_cols, cat_but_car = grab_col_names(df)

    Uyarı:
        cat_cols + num_cols + cat_but_car toplamı, toplam sütun sayısını verir.
    """

    # 1. Kategorik sütunların isimlerini bul ("object" tipinde olanlar)
    cat_cols = [col for col in dataframe.columns if dataframe[col].dtypes == "O"]

    # 2. Numerik görünümlü olup aslında kategorik olan sütunları ekle ("num_but_cat")
    # Yani, veri tipi sayısal olup eşsiz değer sayısı cat_th'den azsa, bunlar da kategori sayılır
    num_but_cat = [col for col in dataframe.columns if dataframe[col].nunique() < cat_th and dataframe[col].dtypes != "O"]

    # 3. Yazısal tipte olup çok fazla eşsiz değeri olan sütunlar ("cat_but_car": kategorik görünümlü kardinal)
    cat_but_car = [col for col in dataframe.columns if dataframe[col].nunique() > car_th and dataframe[col].dtypes == "O"]

    # 4. Kategorik sütunlar listesine numerik tipte olup az kategoriye sahip sütunları ekle
    cat_cols = cat_cols + num_but_cat

    # Ama kategorik görünümlü kardinal (çok değerli) olanları kategoriklerden çıkar
    cat_cols = [col for col in cat_cols if col not in cat_but_car]

    # 5. Numerik sütunların isimlerini bul (tip olarak numerik olanlar)
    num_cols = [col for col in dataframe.columns if dataframe[col].dtypes != "O"]

    # Numerik olup az kategoriye sahip olanlar zaten kategoriklere eklendiği için burada hariç tut
    num_cols = [col for col in num_cols if col not in num_but_cat]

    # Sonucu ekrana yaz (kontrol amaçlı)
    print(f"Gözlem (satır) sayısı: {dataframe.shape[0]}")
    print(f"Değişken (sütun) sayısı: {dataframe.shape[1]}")
    print(f'cat_cols (Kategorik sütun sayısı): {len(cat_cols)}')
    print(f'num_cols (Numerik sütun sayısı): {len(num_cols)}')
    print(f'cat_but_car (Kardinal kategorik sütun sayısı): {len(cat_but_car)}')
    print(f'num_but_cat (Sayısal olup az kategoriye sahip olanlar): {len(num_but_cat)}')
    
    # Üç tür sütun listesini döndür
    return cat_cols, num_cols, cat_but_car
````

````python
# Aykırı Değerlerin Kendilerine Erişmek
def grab_outliers(dataframe, col_name, index=False):
    """
    Belirtilen değişkenin (col_name) aykırı değerlerini veri setinden bulup gösterir.

    Parametreler
    ----------
    dataframe : pd.DataFrame
        İncelenecek veri seti.
    col_name : str
        Aykırı değerleri bulunacak değişkenin ismi.
    index : bool, optional
        True ise, aykırı gözlem indexlerini döndürür. False ise yalnızca aykırı gözlemleri yazdırır.

    Dönüş
    -------
    None veya pd.Index:
        index=True ise, aykırı gözlem indexlerini döndürür. Aksi halde None döner.

    Açıklama
    -------
    Değişkenin alt ve üst aykırı sınırları outlier_thresholds fonksiyonu ile bulunur.
    Eğer aykırı değer sayısı 10'dan fazlaysa, sadece ilk 5'i gösterilir.
    Aykırı değerlerin indexi istenirse, pandas Index nesnesi olarak döndürülür.
    """
    # Değişkenin alt (low) ve üst (up) aykırı değer eşiklerini hesapla
    low, up = outlier_thresholds(dataframe, col_name)

    # Eğer ilgili değişkende aykırı gözlem sayısı 10'dan fazlaysa, ilk 5 aykırı değeri yazdırır
    if dataframe[((dataframe[col_name] < low) | (dataframe[col_name] > up))].shape[0] > 10:
        print(dataframe[((dataframe[col_name] < low) | (dataframe[col_name] > up))].head())
    # Eğer 10 veya daha az aykırı gözlem varsa, hepsini yazdırır
    else:
        print(dataframe[((dataframe[col_name] < low) | (dataframe[col_name] > up))])

    # index parametresi True ise, aykırı değerlerin indekslerini döndürür
    if index:
        outlier_index = dataframe[((dataframe[col_name] < low) | (dataframe[col_name] > up))].index
        return outlier_index
````

````python
def remove_outlier(dataframe, col_name):
    """
    Belirtilen değişkende (col_name) aykırı değerleri kaldırır ve yeni bir DataFrame döndürür.

    Parametreler
    ------------
    dataframe : pd.DataFrame
        Üzerinde aykırı değerlerin aranacağı veri çerçevesi.
    col_name : str
        Aykırı değerlerin aranacağı sütun ismi.

    Dönüş Değeri
    -------------
    pd.DataFrame
        Aykırı değerleri içermeyen yeni veri çerçevesi.

    Açıklama
    ----------
    Bu fonksiyon, outlier_thresholds fonksiyonu ile belirlenen alt ve üst eşik değerlerine göre,
    ilgili sütunda (col_name) eşiklerin dışında kalan (aykırı) gözlemleri veri setinden çıkarır.
    Sonuç olarak, aykırı değerlerden arındırılmış bir DataFrame elde edilir.
    """

    # Alt ve üst aykırı değer eşiklerini belirle
    low_limit, up_limit = outlier_thresholds(dataframe, col_name)

    # Eşiklerin dışında kalan aykırı değerleri filtrele
    df_without_outliers = dataframe[~((dataframe[col_name] < low_limit) | (dataframe[col_name] > up_limit))]

    # Yeni DataFrame'i döndür
    return df_without_outliers
````

````python
def replace_with_thresholds(dataframe, variable):
    """
    Belirtilen değişkenin (variable) aykırı değerlerini alt ve üst sınırlarla değiştirir.

    Parametreler
    ----------
    dataframe : pd.DataFrame
        Aykırı değerleri değiştirilecek veri çerçevesi.
    variable : str
        Değiştirilecek değişkenin ismi.

    Dönüş Değeri
    -------------
    pd.DataFrame
        Aykırı değerleri içermeyen yeni veri çerçevesi.

    Açıklama
    ----------
    Bu fonksiyon, outlier_thresholds fonksiyonu ile belirlenen alt ve üst eşik değerlerine göre,  ilgili sütunda (variable) eşiklerin dışında kalan (aykırı) gözlemleri veri setinden çıkarır.
    Sonuç olarak, aykırı değerlerden arındırılmış bir DataFrame elde edilir.
    """

    # Alt ve üst aykırı değer eşiklerini belirle
    low_limit, up_limit = outlier_thresholds(dataframe, variable)
    
    # Eşiklerin dışında kalan aykırı değerleri filtrele
    dataframe.loc[(dataframe[variable] < low_limit), variable] = low_limit
    dataframe.loc[(dataframe[variable] > up_limit), variable] = up_limit
    
    # Yeni DataFrame'i döndür
    return dataframe

df = load()
df.shape
````

## Eksik Değerler (Missing Values)
Eksik değerler, veri analizinde oldukça sık karşılaşılan ve analizin doğruluğunu etkileyebilen önemli problemlerdendir. Eksik gözlemler, veri setindeki bazı değişkenlerde değerlerin tanımsız veya kayıp olmasından kaynaklanır. Eksik değerlerin doğru şekilde tespit edilmesi ve uygun yöntemlerle işlenmesi analizin güvenilirliğini artırır.

Eksik değerlerin tespiti için aşağıdaki işlemler uygulanabilir:
- Eksik değere sahip olan gözlemlerin ve değişkenlerin bulunması.
- Eksik değerlerin oranlarının hesaplanması ve görselleştirilmesi.
- Eksik değerlerin, veri seti içerisindeki potansiyel etkilerinin incelenmesi.

Eksik değerlerle baş etme yöntemleri:
- İlgili gözlemlerin veya değişkenlerin veri setinden silinmesi (listwise/pairwise deletion).
- Eksik değerlerin uygun istatistiksel yöntemler ile doldurulması (ortalama, medyan, mod, regresyon ile tahmin vb.).
- Eksik değerleri tahmine dayalı yöntemler

Sonuç olarak, eksik değerler analiz sürecinde ihmal edilmemelidir. Eksik değerlerin büyüklüğüne, dağılımına ve nedenine göre uygun yöntem belirlenerek işlenmelidir.

Eksik değere sahip gözlemlerin veri setinden direkt çıkarılması ve rassallığının incelenmemesi, yapılacak istatistiksel çıkarımların ve modelleme çalışmalarının güvenilirliğini düşürecektir. (Alpar, 2011)


Eksik gözlemlerin veri setinden direkt çıkarılabilmesi için, veri setindeki eksikliğin bazı durumlarda kısmen, bazı durumlarda ise tamamen rastlantısal olarak oluşmuş olması gerekmektedir.

Eğer eksiklikler değişkenler ile ilişkili olarak ortaya çıkan yapısal problemler sonucu meydana gelmiş ise, bu durumda yapılacak silme işlemleri ciddi yanlılıklara sebep olabilecektir.
(Tabachnick ve Fidell, 1996)

````python
def missing_values_table(dataframe, na_name=False):
    """
    Dataframe'deki eksik değerlerin analizini yapar ve özet bir tablo halinde gösterir.

    Args:
        dataframe (pd.DataFrame): İncelenecek veri seti.
        na_name (bool): True verilirse yalnızca eksik değere sahip kolon isimlerini döndürür.

    Returns:
        Eksik değere sahip kolon isimleri (opsiyonel).
    """
    # Eksik değere sahip kolonları bulma
    na_columns = [col for col in dataframe.columns if dataframe[col].isnull().sum() > 0]

    # Eksik değer sayısı (n_miss) ve toplam içindeki oranı (ratio) hesaplama
    n_miss = dataframe[na_columns].isnull().sum().sort_values(ascending=False)
    ratio = (dataframe[na_columns].isnull().sum() / dataframe.shape[0] * 100).sort_values(ascending=False)

    # Sonuçları tek tabloda birleştirme
    missing_df = pd.concat([n_miss, np.round(ratio, 2)], axis=1, keys=['Eksik Değer Sayısı', 'Oran (%)'])

    # Tablonun ekrana yazdırılması
    print(missing_df, end="\n")

    # Eğer istenirse, eksik değere sahip kolonların listesini döndürür
    if na_name:
        return na_columns
````

### Eksik Değerlerin Bağımlı Değişken ile İlişkisinin İncelenmesi


````python
# Buradaki kod, eksik değerlerin bağımlı değişken (örneğin hayatta kalma durumu "Survived") ile ilişkisinin analiz edilmesine yöneliktir.

# missing_values_table fonksiyonu ile df veri setindeki eksik verileri tespit ediyoruz ve detaylı tablo elde ediyoruz.
missing_values_table(df, True)
na_cols = missing_values_table(df, True)  # Eksik veri içeren sütunların isim listesi

# Ardından, eksik veriye sahip değişkenlerin bağımlı değişken ile ilişkisini inceleyen bir fonksiyon tanımlanıyor.
def missing_vs_target(dataframe, target, na_columns):
    temp_df = dataframe.copy()  # Orijinal veri setinin bir kopyası alınır.

    # Eksikliği bayraklayan (flag) sütunlar oluşturuluyor:
    for col in na_columns:
        temp_df[col + '_NA_FLAG'] = np.where(temp_df[col].isnull(), 1, 0)

    # Sadece oluşturulan NA flaglerini seçiyoruz:
    na_flags = temp_df.loc[:, temp_df.columns.str.contains("_NA_")].columns

    # Her eksiklik bayrağı için, gruplandırıp bağımlı değişken ortalaması (TARGET_MEAN) ve count (Count) veriliyor.
    for col in na_flags:
        print(pd.DataFrame({
            "TARGET_MEAN": temp_df.groupby(col)[target].mean(),
            "Count": temp_df.groupby(col)[target].count()
        }), end="\n\n\n")

# Fonksiyonu kullanarak, eksikliğin bağımlı değişken üzerindeki etkisini inceliyoruz.
missing_vs_target(df, "Survived", na_cols)
````

### 3. Encoding (Label Encoding, One-Hot Encoding, Rare Encoding)

Makine öğrenmesinde, kategorik değişkenler genellikle doğrudan kullanılmaz. Bu yüzden, kategorik verileri sayısal forma dönüştürmek gerekir. Bu işlem "encoding" olarak adlandırılır. Başlıca encoding yöntemleri şunlardır:

#### Label Encoding
Bu yöntemde, her benzersiz kategori bir sayı ile temsil edilir. Örneğin; "Erkek" = 0, "Kadın" = 1. Genellikle sadece iki farklı değer alan kategorik değişkenlerde tercih edilir.

#### One-Hot Encoding
Her kategori için ayrı bir sütun (kolon) oluşturulur. Gözlem ilgili kategoriye sahipse 1, yoksa 0 yazılır. Böylece herhangi bir sıralama ilişkisi oluşturulmaz ve çok kategorili değişkenlerde kullanılır.

#### Rare Encoding
Veri setinde çok az sayıda gözlemi bulunan, yani nadir kategoriler tek bir ortak kategori altında toplanır. Bu sayede modelin nadir değerlerden etkilenmesi önlenmiş olur.

Bu encoding yöntemleri ile veriler sayısal hale getirildikten sonra makine öğrenmesi algoritmalarında kolaylıkla kullanılabilir.


### Label Encoding

Label Encoding, makine öğrenmesinde kategorik değişkenleri sayısal değerlere dönüştürmek için kullanılan en temel tekniklerden biridir. Özellikle yalnızca iki farklı kategori (örneğin: Evet/Hayır, Erkek/Kadın) içeren değişkenler için uygundur.

#### Nasıl Çalışır?

Her bir kategoriye benzersiz bir tam sayı atanır. İkili kategorili değişkenlerde bu sayılar genellikle 0 ve 1 olur. Bu dönüşüm sayesinde ilgili değişkenler doğrudan modellemeye dahil edilebilir.

### One Hot Encoding 

Bir veri kümesinde, kategori (örneğin şehir, renk, cinsiyet gibi) tipinde değişkenler bulunabilir.  
Makine öğrenmesi algoritmaları ise genellikle sayısal (nümerik) verilerle çalışır. Bu nedenle, kategorik verileri algoritmaların anlayabileceği şekilde sayılara çevirmek gerekir.

**Label Encoding** kullandığımızda, kategorileri sırayla sayılara (ör: 0, 1, 2) dönüştürürüz. Ancak bu yöntem, bazen yanlışlıkla kategoriler arasında bir sıralama/öncelik varmış gibi bir anlam katabilir.

**One Hot Encoding** yöntemiyle ise her kategori için yeni bir sütun (kolon) oluşturulur. Değer, hangi kategoriye aitse sadece o sütunda 1 olur; diğerlerinde 0 olur.

#### Örnek

"Renk" adında 3 kategoriye sahip bir değişkenimiz olsun: `Kırmızı`, `Yeşil`, `Mavi`

**Orijinal veri tablosu:**

|   | Renk    |
|---|---------|
| 0 | Kırmızı |
| 1 | Yeşil   |
| 2 | Mavi    |
| 3 | Kırmızı |

**One Hot Encoding uyguladığımızda tablo şöyle olur:**

|   | Renk_Kırmızı | Renk_Yeşil | Renk_Mavi |
|---|:------------:|:----------:|:---------:|
| 0 |      1       |     0      |     0     |
| 1 |      0       |     1      |     0     |
| 2 |      0       |     0      |     1     |
| 3 |      1       |     0      |     0     |

Yani, her alt tip için ayrı bir sütun oluştu ve ait olduğu yerde 1, diğerlerinde 0 oldu.

Böylece makine öğrenmesi modelleri sayısal olarak kategorik veriyi doğru şekilde ayırt edebilir.

## Rare Encoding Nedir?

Rare encoding, veri kümelerindeki kategorik değişkenlerde nadir (seyrek/az görülen) sınıfların (kategorilerin) tespiti ve yönetimi için kullanılan bir ön işleme yöntemidir. Kategorik değişkende çok fazla sınıf varsa, bazı sınıflar veri setinde çok az sayıda gözleme sahip olabilir. Makine öğrenmesi modelleri, bu nadir görülen sınıfları öğrenmekte zorlanabilir veya bu sınıflar noise (gürültü) yaratabilir.

### Rare Encoding’in Amacı Nedir?
- Nadiren gözlemlenen sınıfları birleştirerek çok sayıda ve dengesiz kategori problemiyle başa çıkmak.
- Modelin daha kararlı, genellenebilir ve önyargısız sonuçlar üretmesini sağlamak.
- Overfitting riskini azaltmak.

### Rare Encoding Nasıl Uygulanır?
1. Belirli bir eşik değerin (örneğin %1, %5) altında kalan sınıflar nadir (rare) olarak kabul edilir.
2. Bu nadir sınıflar, genellikle “Rare” ya da “Other” adında yeni bir kategori altında birleştirilir.
3. Artık her kategorik değişkende, yeterli sayıda gözlemin olduğu ve modelin daha sağlıklı öğrenebileceği bir sınıf dağılımı sağlanır.

### Örnek

Örneğin, “Meslek” sütununda 20 farklı meslek bulunuyor ve bunlardan sadece 3’ü veri setinin %90’ını oluşturuyorsa, kalan az görülen mesleklerin hepsi “Rare” etiketiyle tek bir kategori altında birleştirilebilir.

```python
# Sık olmayan (rare) kategorileri bulmak ve birleştirmek için örnek kod:
def rare_encoder(dataframe, column, rare_perc=0.01):
    freq = dataframe[column].value_counts() / len(dataframe)
    rare_labels = freq[freq < rare_perc].index
    dataframe[column] = dataframe[column].apply(lambda x: "Rare" if x in rare_labels else x)
    return dataframe

# Kullanım:
df = rare_encoder(df, "Meslek", rare_perc=0.01)
```

### Avantajları
- Kategorik değişkenlerdeki noise ve sapmayı azaltır.
- Modelin genelleme başarısını artırır.
- Düşük gözlemli kategorilerin modele olumsuz etkisini sınırlar.

### Dikkat Edilmesi Gerekenler
- Nadir kategorileri birleştirirken domain bilgisinden yararlanmak faydalı olabilir.
- Aşırı birleştirme bilgi kaybına sebep olabilir; eşik değeri dikkatle seçilmelidir.

````python
# rare_analyser fonksiyonu, veri setinde nadir (rare) kategorileri incelemeye yarar.
# Her bir kategorik değişken (cat_cols listesindeki her sütun) için;
#  - O sütunun kaç farklı kategori (sınıf) içerdiğini yazar.
#  - Her bir kategorinin kaç tane (COUNT) olduğu ve veri setindeki oranını (RATIO) gösterir.
#  - Aynı zamanda, her bir kategoriye karşılık hedef değişkene (target) ait ortalama değeri (TARGET_MEAN) gösterir.
# Bu bilgiler tablo halinde yazdırılır. end="\n\n\n" parametresi ise iki satır aralıklı şekilde sonucu ayırır.

def rare_analyser(dataframe, target, cat_cols):
    for col in cat_cols:
        # Sütundaki farklı kategori (unique) sayısını yazdır
        print(col, ":", len(dataframe[col].value_counts()))
        # Her bir kategori için:
        #   COUNT: Adet
        #   RATIO: Toplam içindeki oran
        #   TARGET_MEAN: O kategoriye sahip satırlarda hedef değişkenin ortalaması
        print(
            pd.DataFrame({
                "COUNT": dataframe[col].value_counts(),
                "RATIO": dataframe[col].value_counts() / len(dataframe),
                "TARGET_MEAN": dataframe.groupby(col)[target].mean()
            }),
            end="\n\n\n"
        )

# Fonksiyon çağırımı: Tüm kategorik sütunlar için rare_analyser fonksiyonu kullanılıyor.
rare_analyser(df, "TARGET", cat_cols)
````

````python
# 3. Rare encoder'ın yazılması.
# Rare encoder fonksiyonu ile, veri setinde belirli bir orandan (rare_perc) daha az görülen nadir kategoriler 'Rare' etiketiyle birleştirilir. Bu, kategorik değişkenlerin daha anlamlı ve model için daha yorumlanabilir hale gelmesini sağlar.

def rare_encoder(dataframe, rare_perc):
    # Orijinal veri setini kopyalarız, böylece ana veri etkilenmez.
    temp_df = dataframe.copy()

    # 1. Adım: Nadiren görülen kategori barındıran değişkenleri tespit et.
    #   - Sadece tipik olarak 'O' (object) tipinde olan, yani kategorik olan değişkenlere bakılır.
    #   - Her değişken için dağılım oranı (value_counts() / toplam gözlem sayısı) rare_perc değerinden küçük olanlar nadir (rare) olarak kabul edilir.
    rare_columns = [col for col in temp_df.columns 
                   if temp_df[col].dtypes == 'O'
                   and (temp_df[col].value_counts() / len(temp_df) < rare_perc).any(axis=None)]

    # 2. Adım: Tespit edilen her nadir kategorili sütunda...
    for var in rare_columns:
        # Sütundaki her bir kategorinin oranını hesapla.
        tmp = temp_df[var].value_counts() / len(temp_df)
        # Oranı rare_perc değerinin altında olan kategori isimlerini ('etiketleri') seçiyoruz.
        rare_labels = tmp[tmp < rare_perc].index
        # Bu etiketlere sahip gözlem değerlerini 'Rare' ile değiştiriyoruz.
        temp_df[var] = np.where(temp_df[var].isin(rare_labels), 'Rare', temp_df[var])

    # Fonksiyonun çıktısı, nadir kategoriler 'Rare' etiketiyle birleştirilmiş yeni bir DataFrame'dir.
    return temp_df

# Fonksiyonu uygulayalım:
# Tüm kategorik değişkenlerde %1'den daha az orana sahip kategoriler, 'Rare' olarak birleştirilecek.
new_df = rare_encoder(df, 0.01)

# Yeni veri setinde nadir kategorilerle hedef değişken arasındaki ilişkiyi tekrar analiz ediyoruz.
rare_analyser(new_df, "TARGET", cat_cols)
````

## Feature Scaling (Özellik Ölçeklendirme)

Özellik ölçeklendirme, farklı ölçeklere sahip değişkenlerin makine öğrenmesi modellerinde adil şekilde değerlendirilmesini sağlamak için kullanılan bir tekniktir. Özellikle mesafe tabanlı modellerde (ör. KNN, Lojistik Regresyon, Destek Vektör Makineleri, Yapay Sinir Ağları) değişkenler arasındaki ölçek farkları modelin başarısını olumsuz etkileyebilir.

Aşağıda, en yaygın kullanılan bazı ölçeklendirme yöntemlerine dair kısa açıklamalar ve matematiksel formülleri **Markdown matematik modunda** bulabilirsin:

- **Min-Max Scaling (Normalization):**
    - Her bir özelliği 0 ile 1 arasına sıkıştırır.
    - Formül:
      
      $$
        X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}
      $$
    - Özellikle dağılımı bilinmeyen veya outlier'ların önemli olmadığı veriler için uygundur.

- **Standardization (Z-Score Normalization):**
    - Özelliklerin ortalaması 0, standart sapması 1 olacak şekilde yeniden ölçeklendirir.
    - Formül:

      $$
        X_{stand} = \frac{X - \mu}{\sigma}
      $$
    - Outlier'lara karşı daha dayanıklıdır, dağılımı normal olmayan verilerde de işe yarar.

- **Robust Scaling:**
    - Outlier'ların etkisini azaltmak için medyan ve IQR (interquartile range, çeyrekler arası aralık) kullanır.
    - Formül:

      $$
        X_{robust} = \frac{X - \text{median}}{IQR}
      $$
    - Özellikle veri setinde uç değerlerin çok olduğu durumlarda etkilidir.

Neden Özellik Ölçeklendirme?

- Farklı birimlere (ör. yaş ve gelir) sahip değişkenleri aynı referans çerçevesinde karşılaştırabilmek için gereklidir.
- Özellikle mesafe temelli algoritmalarda (KNN, k-means, PCA vs.) tek bir değişkenin ölçeği, diğerlerini domine edebilir.
- Gradient descent kullanan algoritmalarda daha hızlı ve etkili bir yakınsama sağlar.

Ne Zaman Yapılır?

- Sadece sayısal olan (numerik) sütunlara uygulanır.
- Özellik mühendisliğinden ve veri bölünmesinden (train/test split) sonra, yalnızca eğitim verisine fit edilerek test verisine aynen uygulanmalıdır. Böylece veri sızıntısı (data leakage) engellenmiş olur.

````python
# StandardScaler hem açıklar, hem de uygular: Özelliklerin ortalamasını (mean) 0'a, standart sapmasını (std) 1'e çeker.
# Formül: z = (x - u) / s, burada u: ortalama, s: standart sapma. Böylece her değer kaç standart sapma uzakta gösterilir.
# Bu sayede farklı ölçeklerdeki sayısal değişkenler karşılaştırılabilir hale gelir; özellikle makine öğrenmesi modellerinde önemlidir.

df = load()  # Ham veriyi yükle
ss = StandardScaler()  # Standartlaştırıcıyı oluştur
# Sadece "Age" sütununu standartlaştır ve sonucu yeni bir sütun olarak sakla
df["Age_standard_scaler"] = ss.fit_transform(df[["Age"]])  
df.head()  # İlk birkaç satırı göstererek sonucu kontrol et
````

````python
# RobustScaler, uç değerlerin (outlier) etkisini azaltmak için medyan ve IQR (çeyrekler arası aralık) kullanır. 
# Bu yöntem, özellikle Age (yaş) gibi uç değerlere duyarlı olabilecek değişkenlerde tercih edilir.
# Formül: z = (x - medyan) / IQR
# Böylece, Age sütunu robust bir şekilde ölçeklenmiş olur.

rs = RobustScaler()  # RobustScaler nesnesi oluşturulur
df["Age_robust_scaler"] = rs.fit_transform(df[["Age"]])  # "Age" sütununu robust ölçekle ve yeni sütun olarak ekle
df.describe().T  # Tüm özelliklerin istatistiksel özetine bakılır
````

````python
# MinMaxScaler, veriyi 0 ile 1 arasına sıkıştırır. Formül: (x - min) / (max - min)
# Yani, her bir değeri minimum ve maksimum değere göre ölçekler.
# Özellikle dağılımı bozmak istemediğimiz ve değerlerin belirli bir aralığa girmesi gereken durumlarda tercih edilir.

mms = MinMaxScaler()  # MinMaxScaler nesnesini oluşturuyoruz
df["Age_min_max_scaler"] = mms.fit_transform(df[["Age"]])  # Age sütununu 0-1 arasında ölçekleyip yeni bir sütun olarak ekliyoruz
df.describe().T  # Yeni sütun dahil tüm sayısal değişkenlerin istatistiksel özetini alıyoruz
````

Makine öğrenmesinde "feature extraction" (özellik çıkarımı), ham veriden modelin daha iyi anlamasına ve öğrenmesine yardımcı olacak anlamlı ve bilgilendirici yeni değişkenler/sütunlar oluşturma işlemidir. Yani veri kümesindeki mevcut bilgilerden, modele katkı sağlayacak ek sütunlar üretiriz.

Bu süreçte, mevcut sütunlardaki bilgileri dönüştürerek ya da özetleyerek modelin örüntüleri daha iyi yakalamasını sağlayabiliriz. Örneğin:

- Bir doğum tarihi sütunundan yaş, yaş grubunu ya da doğduğu on yılı çıkarmak
- Metin verisinden metin uzunluğunu, kelime sayısını ya da anahtar kelimeleri çıkarmak
- Kategorik değişkenden anlamlı etiket veya grup sütunları oluşturmak

Feature extraction sayesinde:
- Modelin doğruluğu artabilir,
- Karmaşıklık azalabilir,
- Yoruma açık ve daha anlaşılır değişkenler elde edilebilir.

````python
df["NEW_CABIN_BOOL"] = df["Cabin"].notnull().astype('int')

df.groupby("NEW_CABIN_BOOL").agg({"Survived": "mean"})
````

````python
# Yolcuların Kabin bilgisinin (Cabin sütununun dolu olup olmamasının) hayatta kalma oranı üzerinde anlamlı bir fark yaratıp yaratmadığını test edeceğiz.
# Bunun için "iki oran z-testi" (proportions_ztest) uygulanır. Burada iki grup var:
# - Kabin bilgisi olanlar (NEW_CABIN_BOOL == 1)
# - Kabin bilgisi olmayanlar (NEW_CABIN_BOOL == 0)
# Her iki grupta hayatta kalanların oranlarını karşılaştırıp istatistiksel olarak anlamlı bir fark olup olmadığını inceleyeceğiz.

from statsmodels.stats.proportion import proportions_ztest

# Her iki grupta hayatta kalan kişi sayıları:
survived_with_cabin = df.loc[df["NEW_CABIN_BOOL"] == 1, "Survived"].sum()
survived_without_cabin = df.loc[df["NEW_CABIN_BOOL"] == 0, "Survived"].sum()

# Her iki gruptaki toplam yolcu sayısı:
total_with_cabin = df.loc[df["NEW_CABIN_BOOL"] == 1, "Survived"].shape[0]
total_without_cabin = df.loc[df["NEW_CABIN_BOOL"] == 0, "Survived"].shape[0]

# Z-testi uygulanıyor:
test_stat, pvalue = proportions_ztest(
    count=[survived_with_cabin, survived_without_cabin],
    nobs=[total_with_cabin, total_without_cabin]
)

print(f"Kabin bilgisi olanlar ve olmayanların hayatta kalma oranları arasında istatistiksel olarak anlamlı bir fark var mı?")
print(f"Test İstatistiği (Z): {test_stat:.4f}")
print(f"p-değeri: {pvalue:.4f}")

if pvalue < 0.05:
    print("Sonuç: p < 0.05 olduğu için, iki grup arasında anlamlı bir fark vardır.")
else:
    print("Sonuç: p >= 0.05 olduğu için, iki grup arasında anlamlı bir fark yoktur.")

# Sonuç:
# p-değeri 0.05'ten küçükse (genellikle), gruplar arasında anlamlı bir farklılık olduğu söylenebilir.
````

````python
# Yalnız seyahat edenlerle etmeyenler arasında hayatta kalma oranları arasında istatistiksel olarak anlamlı bir fark var mı, iki oran z-testi ile kontrol edelim.
# "YES" grubu yalnız olanları, "NO" grubu ise yanında yakını bulunan yolcuları ifade eder.
test_stat, pvalue = proportions_ztest(
    count=[
        df.loc[df["NEW_IS_ALONE"] == "YES", "Survived"].sum(),
        df.loc[df["NEW_IS_ALONE"] == "NO", "Survived"].sum()
    ],
    nobs=[
        df.loc[df["NEW_IS_ALONE"] == "YES", "Survived"].shape[0],
        df.loc[df["NEW_IS_ALONE"] == "NO", "Survived"].shape[0]
    ]
)

print("Yalnız olanlar ve olmayanlar arasında hayatta kalma oranı açısından anlamlı fark var mı?")
print("Test Stat = %.4f, p-value = %.4f" % (test_stat, pvalue))
````

````python
# Özellik (feature) önemlerini görselleştirmek için kullanacağımız fonksiyonu tanımlıyoruz.
# Bu fonksiyon, bir modele ve ona ait özelliklere (feature) ait isimleri alır.
# Modelin 'feature_importances_' özelliği ile her bir özelliğin model için ne kadar önemli olduğunu öğreniyoruz.
# İstersek, en önemli ilk 'num' kadar özelliği gösterebiliriz. (Varsayılan olarak tümünü gösterir.)
# Eğer 'save=True' olursa, grafiği kaydeder.

def plot_importance(model, features, num=len(X), save=False):
    # Modelden elde edilen özellik önemlerini ve isimlerini bir DataFrame içinde topluyoruz.
    feature_imp = pd.DataFrame({'Value': model.feature_importances_, 'Feature': features.columns})
    # Grafik boyutunu belirliyoruz.
    plt.figure(figsize=(10, 10))
    # Seaborn grafik ayarını yapıyoruz.
    sns.set(font_scale=1)
    # En önemli özellikleri büyükten küçüğe sıralayarak çubuk grafikle gösteriyoruz.
    sns.barplot(
        x="Value",
        y="Feature",
        data=feature_imp.sort_values(by="Value", ascending=False)[0:num]
    )
    # Grafiğe başlık veriyoruz.
    plt.title('Features')
    # Grafik üzerindeki ögeleri daha düzenli yerleştiriyoruz.
    plt.tight_layout()
    # Grafik ekranda gösteriliyor.
    plt.show()
    # Eğer istenirse, grafiği importances.png olarak kaydediyoruz.
    if save:
        plt.savefig('importances.png')

# Tanımladığımız fonksiyonu çağırarak modelimizin özellik önemlerini görselleştiriyoruz.
plot_importance(rf_model, X_train)
````
