### Item-Based Collaborative Filtering (Nesne-Tabanlı İş Birlikçi Filtreleme)

Item-Based Collaborative Filtering, kullanıcılara daha önce puanladıkları veya beğendikleri ürünlere (film, kitap vb.) benzeyen diğer ürünleri öneren bir öneri sistemidir.

Yöntemin temel adımları şunlardır:

1. **Kullanıcı-Ürün Matrisinin Oluşturulması:** Kullanıcıların ürünlere verdikleri puanlardan oluşan bir matris oluşturulur.
2. **Ürünler Arası Benzerliğin Hesaplanması:** Her ürünün diğer ürünlerle olan benzerliği (örneğin korelasyon, kosinüs benzerliği veya başka bir benzerlik ölçütü ile) hesaplanır.
3. **Önerilerin Sunulması:** Kullanıcının yüksek puan verdiği bir ürün seçilir ve bu ürünle en benzer ürünler öneri olarak sunulur.

**Avantajları:**
- Özellikle çok fazla kullanıcı ve ürün olduğunda, ürün profilleri göreli olarak daha sabit olduğu için genellikle daha tutarlı öneriler sunar.

**Dezavantajları:**
- Çok az puanlanmış ürünlerde öneri yapmak zorlaşabilir.
- Ürün sayısı çok fazla olursa benzerlik matrisini saklamak maliyetli olabilir.

Kullanım alanları genellikle film, müzik, alışveriş ve içerik öneri sistemleridir.

### İş Problemi

Online bir film izleme platformu (örneğin kuzukuzu.tv) iş birlikçi filtreleme yöntemi ile bir öneri sistemi geliştirmek istemektedir.

İçerik temelli öneri sistemlerini deneyen şirket, topluluğun kanaatlerini barındıracak şekilde, yani kullanıcı tabanlı değerlendirmelere dayanan öneriler geliştirmek istemektedir.

Kullanıcılar bir filmi beğendiğinde, ona benzer beğenilme örüntüsüne sahip olan diğer filmlerin önerilmesi hedeflenmektedir.

### Veri Seti Hikayesi

Veri seti MovieLens tarafından sağlanmıştır.

İçerisinde filmler ve bu filmlere verilen puanlar bulunmaktadır.

Veri seti yaklaşık 27.000 film ve yaklaşık 2.000.000 derecelendirme içermektedir.

Veri seti: https://grouplens.org/datasets/movielens/

**movie.csv**

- `movieId`: Eşsiz film numarası (benzersiz film ID'si)
- `title`: Film adı

**rating.csv**

- `userId`: Eşsiz kullanıcı numarası (benzersiz kullanıcı ID'si)
- `movieId`: Eşsiz film numarası (benzersiz film ID'si)
- `rating`: Kullanıcı tarafından filme verilen puan/değerlendirme
- `timestamp`: Değerlendirme tarihi (zaman damgası)

### 1. Veri Setinin Hazırlanması

```python
import pandas as pd
pd.set_option('display.max_columns', 500)
movie = pd.read_csv('datasets/movie_lens_dataset/movie.csv')
rating = pd.read_csv('datasets/movie_lens_dataset/rating.csv')
# movie ve rating veri setleri, "movieId" değişkenine göre birleştirilerek (left join) tek bir veri seti olan df oluşturuluyor.
df = movie.merge(rating, how="left", on="movieId")
df.head()
```
