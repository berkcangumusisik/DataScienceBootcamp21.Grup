## Metin Üretme ve Sohbet Tabanlı Yapay Zeka Arasındaki Temel Farklar

Bu görsel, "Metin Üretme (Text Generation)" ve "Sohbet/Conversational AI" yaklaşımlarını karşılaştırmaktadır.

### Completion/Generation (Metin Üretme)
- **İstek (Prompt):** Metin üretmek için modelin başlangıç verisi olarak kullanılır.
- **Context Window:** Modelin aynı anda dikkate alabildiği maksimum token (kelime/parça) sayısıdır.
- **Temperature & Maximum Tokens:** Temperature, üretilen metnin rastgeleliğini/yaratıcılığını kontrol eder; Maximum Tokens ise oluşturulacak cevabın maksimum uzunluğunu sınırlar.
- **Step Sequence:** Bazı modellerde metin üretimi birkaç aşamada yapılabilir.
- **Instruction-Based:** Model, kendisine verilen talimatları takip ederek istenilen biçimde metin üretebilir.

### Chat/Conversational AI (Sohbet Tabanlı Yapay Zeka)
- **Soru-Cevap ve Önceki Konuşmalar:** Model, önceki konuşmaları dikkate alarak cevap üretir; bu, diyaloğun daha tutarlı ve doğal ilerlemesini sağlar.
- **Aynı Teknik Parametreler:** Context Window, Temperature, Maximum Tokens ve Stop Sequence gibi parametreler burada da kullanılır.
- **System Message:** Modelin rolünü belirleyen mesajlardır. Örneğin, modelin bir öğretmen, müşteri temsilcisi ya da farklı bir karakter gibi davranması sağlanabilir.
