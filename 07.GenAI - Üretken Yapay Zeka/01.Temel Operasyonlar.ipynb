## Üretken Yapay Zeka vs Klasik Yapay Zeka

**Klasik yapay zeka** (veya dar yapay zeka), insan zekasına benzer şekilde belirli görevleri yerine getirmek için kurallara dayalı algoritmalar ve mantıksal çıkarım mekanizmaları kullanır. Bu sistemler, önceden belirlenmiş kurallar veya veriler üzerinden karar verir; örneğin, satranç oynayan bir yapay zeka veya e-posta filtresi bu tipe örnektir. Klasik yapay zekada karar verme süreci daha öngörülebilir ve şeffaftır.

**Üretken yapay zeka** ise (Generative AI), yeni ve özgün veriler üretebilen gelişmiş modellerden oluşur. Özellikle sinir ağları ve derin öğrenme mimarilerini kullanır. Resim oluşturabilen, metin yazabilen veya müzik besteleyebilen modeller bu kategoriye girer (örneğin ChatGPT, DALL-E, Midjourney). Üretken yapay zeka, mevcut büyük veri kümelerinden öğrenerek benzeri görülmemiş içerikler üretebilir.

**Temel farklar:**
- *Klasik Yapay Zeka*: Belirli kurallar ve mantıkla sınırlı; el ile programlanan çözümler; üretici değil, karar vericidir.
- *Üretken Yapay Zeka*: Öğrenilmiş örüntülere göre yeni veri ve içerik üretebilir; yaratıcıdır ve insan benzeri çıktı üretebilir.

**Kısaca:**  
Klasik yapay zeka var olan veriler üzerinde analiz ve karar verirken, üretken yapay zeka sıfırdan yeni ve özgün içerikler oluşturur.

## Çekişmeli Üretici Ağlar (Generative Adversarial Networks - GANs)

**Çekişmeli Üretici Ağlar (GANs)**, 2014 yılında Ian Goodfellow ve arkadaşları tarafından geliştirilen bir derin öğrenme mimarisidir. GAN'lar iki temel bileşenden oluşur:

- **Üretici (Generator):** Rastgele gürültü ya da belirli bir dağılımdan yeni veri üretmeye çalışır. Amacı, sahte (sentetik) veriler oluşturmaktır.
- **Ayrıştırıcı (Discriminator):** Gerçek veri ile üreticinin oluşturduğu sahte verileri ayırt etmeye çalışan bir ağdır. Amacı, hangi verinin gerçek, hangisinin sahte olduğunu tespit etmektir.

Bu iki ağ, bir oyun şeklinde birbirine karşı çalışır:

1. Üretici, ayrıştırıcıyı kandıracak kadar gerçekçi veriler üretmeye çalışır.
2. Ayrıştırıcı ise gerçek ve sahte verileri ayırt etmeye çalışır.

### Eğitim Süreci

- Başlangıçta üreticinin ürettiği veriler gerçek verilere göre kolayca ayırt edilir.
- Zamanla üretici, daha gerçekçi veriler üretmeyi öğrenir; ayrıştırıcı, üreticinin sahte verilerini ayırt etmekte zorlanmaya başlar.
- Eğitim devam ettikçe üretici ağ, veri dağılımını öğrenir ve neredeyse gerçek kadar kaliteli yeni örnekler oluşturabilir.

### Kullanım Alanları

- Gerçekçi görüntü, ses veya metin üretimi
- Fotoğraflarda çözünürlük artırma (Super Resolution)
- Görüntüden görüntüye çeviriler (ör: siyah-beyazdan renklendirme)
- Veri tabanlı sanat üretimi
- Derin sahte (Deepfake) video/görüntü üretimi

### Avantajlar ve Zorluklar

- GAN'lar, gerçekçi içerik oluşturma konusunda oldukça başarılıdır.
- Ancak eğitimleri zordur ve "mode collapse" gibi sorunlar görülebilir (üretici ağ sadece belirli bir çıktıyı tekrar tekrar üretebilir).

**Özet:**  
Çekişmeli Üretici Ağlar, iki sinir ağının birbirine karşı "yarıştığı" ve bu sayede gerçek veriye çok yakın yeni veriler üretebilen güçlü üretken yapay zeka modelleridir.

## Transformer Mimarisi: Adım Adım Basit ve Anlaşılır Açıklama

Transformer, günümüzde yapay zekanın özellikle dil ve metin işleme alanında kullanılan modern bir yapısıdır. Hiç bilmeyen biri için, temel bileşenleri ve nasıl çalıştığı aşağıda basitçe açıklanmıştır.

---

### 1. Girdi (Input) ve Temsil Edilmesi

Bir cümle ya da metin, önce küçük parçalara ayrılır. Bu parçalara "token" adı verilir. Örneğin:  
"Ben okula gidiyorum." → ["Ben", "okula", "gidiyorum", "."]

Daha sonra her bir token, bilgisayarın anlayacağı şekilde sayı dizilerine (vektörlere) dönüştürülür. Buna "gömme" (embedding) denir.  
Her kelimenin kendine ait bir sayısal ifadesi olur.

---

### 2. Pozisyon Bilgisi (Positional Encoding)

Transformer, kelimelerin diziliş sırasını doğrudan bilmez. Ancak cümledeki kelime sırası önemlidir.  
Bunun için, her kelimeye sırasını belirten ek bilgiler de eklenir.  
Böylece model, hangi kelimenin kaçıncı sırada olduğunu anlar.

---

### 3. Encoder ve Decoder

Transformer iki bölüme ayrılır:

- **Encoder (Kodlayıcı):** Girilen metni işler ve içeriğini özetler.  
- **Decoder (Çözücü):** Encoder’dan aldığı özet bilgilerle yeni bir metin üretir.

Bazı uygulamalarda sadece encoder veya sadece decoder kullanılır.  
(Google Translate gibi çeviri sistemlerinde her ikisi de kullanılırken, bazı modellerde sadece biri olabilir.)

---

### 4. Çoklu Dikkat Mekanizması (Multi-Head Attention)

Bu bölümde, model cümlenin her bir kelimesinin, diğer tüm kelimelerle olan ilişkisini “dikkat” yöntemiyle inceler. Böylece, önemli kelimelere daha çok odaklanır.

Çoklu başlık (multi-head), bu dikkat mekanizmasını aynı anda farklı şekillerde ve farklı açılardan uygular.  
Bu sayede, kelimeler arasındaki karmaşık ilişkiler daha iyi anlaşılır.

---

### 5. Beslemeli İleri Katman (Feed Forward Layer)

Dikkat mekanizmasından sonra, bilgiler basit bir sinir ağına gönderilir.  
Burada bilgiler daha fazla işlenir ve dönüşür.  
Bu katman, kelimelerin anlam bakımından zenginleşmesini sağlar.

---

### 6. Artık Bağlantılar ve Katman Normalizasyonu

Her aşamada bilgi kaybını önlemek için, giriş verileri çıkışa tekrar eklenir. Buna “artık bağlantı” denir.  
Ayrıca, öğrenmenin daha kararlı olması için “katman normalizasyonu” uygulanır.

---

### 7. Katmanlar ve Tekrarlama

Yukarıda anlatılan dikkat ve ileri besleme katmanları, transformer içerisinde genellikle defalarca üst üste sıralanır.  
Her yeni katman, bilgiyi biraz daha “anlamlı” hale getirir.

---

### 8. Sonuç (Çıktı)

Modelin çıkışında, amacı doğrultusunda yeni bir metin üretilir (örneğin: çeviri, özetleme veya soruya yanıt verme).

---

## Bileşenler Nasıl Birbirine Bağlı?

1. **Girdi + Pozisyon Bilgisi**
2. Encoder blokları (Çoklu Dikkat → Beslemeli İleri Katman → Artık Bağlantı ve Normalizasyon)
3. Eğer varsa: Decoder blokları (benzer yapıda)
4. Sonuç üretimi (yeni metin, sınıflandırma vs.)

---

### Kısa Özet

Transformer, metni anlamak ve yeni metinler üretmek için  
- kelimeleri vektörlere çevirir,  
- sıra bilgisini ekler,  
- çoklu dikkat ile ilişkileri inceler,  
- anlamı güçlendirir,  
- aynı işlemi birçok katmanda tekrarlar  
ve sonunda yeni, tutarlı sonuçlar ortaya çıkarır.

Bu yapı sayesinde transformerlar, günümüzde çok başarılı yapay zeka uygulamalarının temelini oluşturuyor.

## Büyük Dil Modelleri (LLM’ler) Nedir? Nasıl Çalışır?

Büyük Dil Modelleri (LLM – Large Language Model), devasa miktarda metin verisi üzerinde eğitilen; insan dilini anlayabilen, analiz edebilen ve yeni metinler üretebilen yapay zeka modelleridir. Günümüzde en bilinen örnekleri GPT (OpenAI), PaLM (Google), LLaMA (Meta) ve benzeri modellerdir.

---

### 1. Temel Bileşenler

- **Veri Kümesi:** LLM’ler, internet üzerinden toplanmış kitaplar, makaleler, diyaloglar, Wikipedia ve kod örnekleri gibi çok büyük ve çeşitli metinlerle eğitilir.
- **Parametreler:** Modelin öğrendiği ağırlıklardır. Modern dil modellerinde, parametre sayısı milyarlarla ifade edilir.
- **Katmanlar:** Bilginin işlendiği transformer bloklarından oluşur. Derinlik arttıkça modelin karmaşıklığı ve ifade gücü de artar.

---

### 2. Eğitim Süreci Nasıl Çalışır?

1. **Veri Hazırlama:** Ham metinler temizlenir, tokenize edilir (kelime ya da alt-kelimelere bölünür) ve vektörlere dönüştürülür.
2. **Dil Modelleme Görevleri:** Model, bir cümle ya da paragrafın devamını tahmin etmek için eğitilir (örnek: “Akşam olunca …” → “hava serinledi.”).
3. **Kayıp Fonksiyonu:** Modelin tahminiyle gerçek cevap arasındaki fark ölçülür.
4. **Geri Yayılım ve Optimizasyon:** Hata minimize edilene kadar ağın ağırlıkları güncellenir. Çok güçlü donanımlarla haftalarca, aylarca sürebilir.

---

### 3. Transformer Mimarisi

Büyük dil modellerinin kalbinde “Transformer” adı verilen bir yapay sinir ağı mimarisi bulunur.

- **Girdi Temsili:** Metin topluluğu, embedding adı verilen çok boyutlu vektörlere dönüştürülür.
- **Pozisyonel Kodlama:** Kelimelerin sırası korunur; böylece model dizideki bağlamı anlayabilir.
- **Çoklu Kendine Dikkat (Self-Attention):** Her kelime, dizinin diğer tüm kelimeleriyle karşılıklı bağını sürekli olarak değerlendirerek bağlamı öğrenir.
- **İleri Beslemeli Katmanlar ve Normalizasyon:** Bilgi bu bloklarda işlendikten sonra daha tutarlı öğrenme için normalizasyon uygulanır.
- **Çok Katmanlı İşleme:** Yukarıdaki yapı, üst üste birçok kez tekrarlanır ve böylece karmaşık dil kalıpları öğrenilir.

---

### 4. Yetenekler ve Kullanım Alanları

- **Dil Anlama ve Üretme:** Soruları yanıtlayabilir, özet çıkarabilir, çeviri yapabilir veya yeni metin oluşturabilir.
- **Kod Yazma ve Hata Ayıklama:** Yazılım geliştiricilere yardımcı olabilir.
- **Sohbet ve Etkileşim:** Kullanıcılarla doğal insan sohbeti gerçekleştirebilir.
- **Veri Analizi & Bilgi Çıkarma:** Büyük veri kümelerinde özetleme veya bilgi çıkarımı yapabilir.

---

### 5. Güçlü ve Zayıf Yönleri

**Avantajlar:**
- Son derece geniş bilgi tabanı
- Çok çeşitli görevlerde güçlü performans
- Doğal ve tutarlı metin üretimi

**Zorluklar:**
- Çok yüksek donanım ve enerji gereksinimi
- Eğitildiği veriden önyargı veya hata öğrenme riski (hallüsinasyon, yanlış bilgi)
- Şeffaflık ve açıklanabilirlik eksikliği

---

### 6. LLM’lerin Önemi

- Çok dilli ve bağlama duyarlı yanıtlar üretebilirler.
- İnsan benzeri etkileşim sunarak birçok sektörde yenilik sağlarlar (eğitim, sağlık, müşteri hizmetleri, içerik üretimi, araştırma vs.).
- Transfer öğrenme sayesinde yeni görevlere çok az veya hiç ek veri ile uyum sağlayabilirler.

---

## Özet

Büyük Dil Modelleri, dilin derin bağlamını ve yapısını anlamak için milyonlarca metin üzerinde eğitilmiş, insanla doğal ve üretken şekilde etkileşime girebilen yapay zekalardır. Modern LLM’ler, çok çeşitli metin tabanlı görevleri başarıyla ve yüksek kalitede otomatikleştirmişlerdir; böylece yapay zeka araştırmasının ve uygulamasının önünü açmışlardır.

**PROMPT | GİRDİ:** Modelin yanıt vermesi için sağlanan metindir. Soru, yönerge ya da örnek içerebilir.

**COMPLETION | ÇIKTI:** Modelin prompt'a karşılık olarak ürettiği yanıttır.

**TOKEN | JETON:** Metnin model tarafından işlenen en küçük birimidir. Bir kelime, kelime parçası veya noktalama işareti olabilir. 1 kelime yaklaşık 1-3 token'a denk gelebilir.

**STREAMING | AKIŞ:** Model yanıtının parça parça, anlık olarak iletilmesidir. Böylece uzun yanıtların tamamını beklemeden görüntüleme yapılabilir.

**PARAMETER | PARAMETRE:** Modelin eğitimi sırasında belirlenen ağırlıklar, ayarlar ve modelin toplam doğruluk ve kapasitesini etkileyen değerlerdir.

**KNOWLEDGE CUTOFF | SON GÜNCEL BİLGİ TARİHİ:** Modelin eğitildiği verilerin son güncellendiği zamandır. Bu tarihten sonraki bilgiler modele dahil değildir.

**CONTEXT WINDOW | BAĞLAM PENCERESİ:** Modelin aynı anda işleyebildiği toplam token (kelime/karakter) miktarıdır. Bu pencere aşıldığında eski bilgiler unutulabilir.

**MULTI-MODALITY | ÇOKLU FORM:** Modelin yalnızca metin değil, görsel, ses veya tablo gibi farklı veri türleriyle de çalışabilmesidir.

**HALLUCINATION | HALLÜSİNASYON:** Modelin, gerçekte olmayan ya da yanlış bilgiler içeren, özgüvenli yanıtlar üretmesidir.

Bağlam penceresi, bir yapay zeka dil modelinin aynı anda işleyebildiği toplam token (kelime veya karakter parçaları) miktarını ifade eder. Model, bir girdiye yanıt üretirken bu pencere içerisindeki verileri kullanır. Eğer metin bu sınırı aşarsa, en eski bilgiler dışarıda kalır ve model yanıt üretirken göz önüne alınmaz. Daha geniş bir bağlam penceresi, modelin daha fazla geçmiş bilgiyi kullanarak daha anlamlı ve tutarlı yanıtlar verebilmesini sağlar.

Büyük dil modellerinde ölçekleme ilkeleri, model boyutu (örneğin katman veya parametre sayısı), eğitim verisi miktarı ve hesaplama kaynaklarının artırılmasıyla modelin doğruluğu, kapasitesi ve öğrenme yeteneği arasındaki ilişkiyi açıklayan prensiplerdir. Temel olarak:

- Parametre ölçeği arttıkça model daha karmaşık ilişkileri öğrenebilir, fakat daha fazla veri ve hesaplama kaynağına ihtiyaç duyar.
- Eğitimde kullanılan veri miktarı büyüdükçe, modelin genelleme yeteneği genellikle güçlenir.
- Büyük modellerin eğitimi için güçlü donanım çözümleri (GPU, TPU, dağıtık sistemler) gerekir.
- Model paralelleştirme, mikrobatch’leme ve checkpointing gibi yöntemlerle eğitim süreci verimli ve ölçeklenebilir hâle getirilir.

Sonuç olarak, doğru ölçekleme stratejileriyle büyük dil modelleri yüksek performansa ulaşabilir; fakat bunun bir maliyeti, sürdürülebilirlik ve verimlilik açısından dikkate alınması gereken sınırları bulunur.
